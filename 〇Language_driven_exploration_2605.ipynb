```python
# ===============================================================
# Complete code:
# BLOX original + BO leak-free version
# BO for Dynamics × Meso-structure
#
# Changes:
#   - Figure labels: ParetoBO -> BO
#   - Figure labels: Text -> Languange
#   - Occupied fraction bar Y ticks: 0, 0.2, 0.4
#   - 2D plot X/Y ticks: 0, 0.5, 1.0
# ===============================================================

!pip -q install scikit-learn scipy matplotlib pandas numpy openpyxl python-pptx

import os
import glob
import re
import zipfile
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import Matern, WhiteKernel, ConstantKernel
from sklearn.ensemble import GradientBoostingRegressor
from scipy.spatial.distance import cdist
from scipy.spatial import ConvexHull

from pptx import Presentation
from pptx.util import Inches, Pt
from pptx.enum.text import PP_ALIGN

warnings.filterwarnings("ignore")

# ===============================================================
# 0. Settings
# ===============================================================

OUTDIR = "./BO_Dynamics_Mesostructure_output"
os.makedirs(OUTDIR, exist_ok=True)

ZIP_PATH  = "./BO_Dynamics_Mesostructure_output.zip"
PPTX_PATH = "./BO_Dynamics_Mesostructure_output.pptx"

N_REPEATS  = 5
BO_ITERS   = 40
BLOX_ITERS = 40

N_POOL = 800
N_CAND = 600

GRID_BINS = (10, 10)
TOTAL_ACTIVE_DEFAULT = 0.97
DYNAMICS_SOURCE_VARIANT = "snd"

NUMERIC_WEIGHT = 1.0
TEXT_WEIGHT = 1.0

MIN_COMPOSITION_DISTANCE = 0.03

TARGET_VARIANTS = [
    "n", "sn", "snc", "snm", "sni", "snd", "snp", "sncmidp"
]

VARIANT_LABEL_MAP = {
    "n": "N",
    "sn": "N+S",
    "snc": "N+SC",
    "snm": "N+SM",
    "sni": "N+SI",
    "snd": "N+SD",
    "snp": "N+SP",
    "sncmidp": "N+SCMIDP",
}

METHODS = [
    "BO-Numerical",
    "BO-N+Languange",
    "BLOX-Numerical",
    "BLOX-N+Languange",
]

METHOD_COLORS = {
    "BO-Numerical": "#4D4D4D",
    "BO-N+Languange": "#0072B2",
    "BLOX-Numerical": "#009E73",
    "BLOX-N+Languange": "#E69F00",
}

METHOD_MARKERS = {
    "BO-Numerical": "o",
    "BO-N+Languange": "s",
    "BLOX-Numerical": "^",
    "BLOX-N+Languange": "D",
}

HYDROPHILIC = ["AMPS", "pSSA", "MEDSAH", "VBA", "NIPAM", "HEA", "DMAA", "AAm"]
HYDROPHOBIC = ["HMA", "TFEMA", "HA"]
CROSSLINKER = ["TECL", "TRCL", "DICL"]
ALL_MONOMERS = sorted(set(HYDROPHILIC + HYDROPHOBIC + CROSSLINKER))

NUMERIC_REPR_COLS_BLOX = [
    "comp_1", "comp_2", "comp_3",
    "headgroup_score_norm",
    "glycerol_score_norm",
    "alkenyl_score_norm",
    "alkyl_chain_score_norm",
    "motif_score_max",
    "motif_breadth_count_0p5",
    "motif_selectivity_index",
    "Peak1_T2_ms",
    "Peak2_T2_ms",
    "Peak3_T2_ms",
    "Peak4_T2_ms",
    "Width_log10T2",
    "Weighted_logmean_T2_ms",
    "N_detected_peaks",
    "Fit_R2",
    "Ridge_alpha",
]

BO_LEAKAGE_COLS = [
    "Peak1_T2_ms",
    "Peak2_T2_ms",
    "Peak3_T2_ms",
    "Peak4_T2_ms",
    "Width_log10T2",
    "Weighted_logmean_T2_ms",
    "N_detected_peaks",
    "Fit_R2",
    "Ridge_alpha",
    "headgroup_score_norm",
    "glycerol_score_norm",
    "alkenyl_score_norm",
    "alkyl_chain_score_norm",
    "motif_score_max",
    "motif_breadth_count_0p5",
    "motif_selectivity_index",
]

NUMERIC_REPR_COLS_BO = [
    c for c in NUMERIC_REPR_COLS_BLOX
    if c not in BO_LEAKAGE_COLS
]

print("BLOX numeric columns:", NUMERIC_REPR_COLS_BLOX)
print("BO leak-free numeric columns:", NUMERIC_REPR_COLS_BO)
print("BO removed leakage columns:", BO_LEAKAGE_COLS)

plt.rcParams["font.family"] = "sans-serif"
plt.rcParams["font.sans-serif"] = ["Arial", "Helvetica", "DejaVu Sans"]
plt.rcParams["figure.dpi"] = 300
plt.rcParams["savefig.dpi"] = 300

LABEL_FS = 24
TICK_FS  = 17
TITLE_FS = 24
LEGEND_FS = 11

# ===============================================================
# 1. File search
# ===============================================================

def find_file(patterns):
    hits = []
    for p in patterns:
        hits.extend(glob.glob(p, recursive=True))
    hits = sorted(set(hits), key=lambda x: os.path.getmtime(x), reverse=True)
    return hits[0] if hits else None

NUMERIC_CSV = find_file([
    "./**/numeric_matrix.csv",
    "/content/**/numeric_matrix.csv",
    "/mnt/data/**/numeric_matrix.csv",
])

TEXT_CSV = find_file([
    "./**/WordToken_features_concept_only.csv",
    "./**/WordToken_features_concept_only*.csv",
    "/content/**/WordToken_features_concept_only.csv",
    "/content/**/WordToken_features_concept_only*.csv",
    "/mnt/data/**/WordToken_features_concept_only.csv",
    "/mnt/data/**/WordToken_features_concept_only*.csv",
])

if NUMERIC_CSV is None or TEXT_CSV is None:
    from google.colab import files
    print("Please upload numeric_matrix.csv and WordToken_features_concept_only.csv")
    files.upload()

    NUMERIC_CSV = find_file([
        "./**/numeric_matrix.csv",
        "/content/**/numeric_matrix.csv",
        "/mnt/data/**/numeric_matrix.csv",
    ])

    TEXT_CSV = find_file([
        "./**/WordToken_features_concept_only.csv",
        "./**/WordToken_features_concept_only*.csv",
        "/content/**/WordToken_features_concept_only.csv",
        "/content/**/WordToken_features_concept_only*.csv",
        "/mnt/data/**/WordToken_features_concept_only.csv",
        "/mnt/data/**/WordToken_features_concept_only*.csv",
    ])

if NUMERIC_CSV is None:
    raise FileNotFoundError("numeric_matrix.csv not found.")
if TEXT_CSV is None:
    raise FileNotFoundError("WordToken_features_concept_only.csv not found.")

print("NUMERIC_CSV:", NUMERIC_CSV)
print("TEXT_CSV   :", TEXT_CSV)

# ===============================================================
# 2. Load data
# ===============================================================

numeric_df = pd.read_csv(NUMERIC_CSV, encoding="utf-8-sig")
text_df = pd.read_csv(TEXT_CSV, encoding="utf-8-sig")

META_COLS = ["row_id", "Copolymer_Name", "variant"]

for c in META_COLS:
    if c not in numeric_df.columns:
        raise KeyError(f"Missing column in numeric_matrix.csv: {c}")
    if c not in text_df.columns:
        raise KeyError(f"Missing column in WordToken_features_concept_only.csv: {c}")

df = numeric_df.merge(
    text_df,
    on=META_COLS,
    how="inner",
    suffixes=("", "_textdup")
)

TEXT_FEATURE_COLS = [
    c for c in text_df.columns
    if c.startswith("TOK_") or c.startswith("SP_")
]

if len(TEXT_FEATURE_COLS) == 0:
    raise ValueError("No TOK_ or SP_ feature columns found.")

for c in sorted(set(NUMERIC_REPR_COLS_BLOX + NUMERIC_REPR_COLS_BO + BO_LEAKAGE_COLS)):
    if c not in df.columns:
        df[c] = np.nan
    df[c] = pd.to_numeric(df[c], errors="coerce").replace([np.inf, -np.inf], np.nan)

print("Loaded:")
print("  numeric_df:", numeric_df.shape)
print("  text_df   :", text_df.shape)
print("  merged    :", df.shape)
print("  text feature cols:", len(TEXT_FEATURE_COLS))

# ===============================================================
# 3. Helper functions
# ===============================================================

def normalize_name(x):
    if pd.isna(x):
        return ""

    s = str(x).strip()
    s = re.sub(r"\(.*?\)", "", s)
    s = s.replace("-", "_").replace("/", "_")
    s = re.sub(r"\s+", "", s)

    s = s.replace("4VBA", "VBA")
    s = s.replace("MEDSH", "MEDSAH")
    s = s.replace("TECL2", "TECL")

    parts = s.split("_")
    if len(parts) >= 3:
        parts = parts[:3]

    repl = {
        "PSSA": "pSSA",
        "AAM": "AAm",
        "AMPS": "AMPS",
        "NIPAM": "NIPAM",
        "HMA": "HMA",
        "TFEMA": "TFEMA",
        "HA": "HA",
        "TECL": "TECL",
        "TRCL": "TRCL",
        "DICL": "DICL",
        "VBA": "VBA",
        "HEA": "HEA",
        "DMAA": "DMAA",
        "MEDSAH": "MEDSAH",
    }

    return "_".join([repl.get(p.upper(), p) for p in parts])

def normalize_variant(v):
    if pd.isna(v):
        return ""
    return str(v).strip().lower().replace("cond_", "").replace("+", "")

def parse_monomer_name(name):
    parts = str(name).split("_")
    if len(parts) < 3:
        return None
    return parts[:3]

def monomer_fraction_vector(name, comp1, comp2, comp3, monomer_list):
    vec = np.zeros(len(monomer_list), dtype=float)
    mons = parse_monomer_name(name)

    if mons is None:
        return vec

    comps = [comp1, comp2, comp3]
    idx_map = {m: i for i, m in enumerate(monomer_list)}

    for m, v in zip(mons, comps):
        if m in idx_map and pd.notna(v):
            vec[idx_map[m]] += float(v)

    return vec

def build_observed_composition_matrix(df_sub, monomer_list):
    rows = []

    for _, r in df_sub.iterrows():
        rows.append(
            monomer_fraction_vector(
                r["Copolymer_Name_norm"],
                r.get("comp_1", np.nan),
                r.get("comp_2", np.nan),
                r.get("comp_3", np.nan),
                monomer_list
            )
        )

    if len(rows) == 0:
        return np.zeros((0, len(monomer_list)))

    return np.vstack(rows)

def scale01(arr):
    arr = np.asarray(arr, float)
    finite = np.isfinite(arr)
    out = np.full_like(arr, np.nan, dtype=float)

    if finite.sum() == 0:
        return out

    mn, mx = np.nanmin(arr[finite]), np.nanmax(arr[finite])

    if mx == mn:
        out[finite] = 0.0
        return out

    out[finite] = (arr[finite] - mn) / (mx - mn)
    return out

def minmax01(x):
    x = np.asarray(x, dtype=float)
    finite = np.isfinite(x)
    out = np.zeros_like(x, dtype=float)

    if finite.sum() == 0:
        return out

    mn = np.nanmin(x[finite])
    mx = np.nanmax(x[finite])

    if mx <= mn:
        out[finite] = 0.5
    else:
        out[finite] = (x[finite] - mn) / (mx - mn)

    return out

def hydrophobic_from_X(X, monomer_list, total_active):
    idx = {m: i for i, m in enumerate(monomer_list)}
    hcols = [idx[m] for m in HYDROPHOBIC if m in idx]

    if len(hcols) == 0:
        return np.zeros(X.shape[0])

    return X[:, hcols].sum(axis=1) / total_active

def hydrophobic_fraction_from_name_and_comps(
    name,
    comp1,
    comp2,
    comp3,
    total_active=TOTAL_ACTIVE_DEFAULT
):
    mons = parse_monomer_name(name)

    if mons is None:
        return np.nan

    comps = [comp1, comp2, comp3]
    h = 0.0

    for m, v in zip(mons, comps):
        if m in HYDROPHOBIC and pd.notna(v):
            h += float(v)

    denom = total_active if total_active > 0 else np.nansum(comps)

    if denom <= 0:
        return np.nan

    return h / denom

def compute_ilt_dynamics_score_row(r):
    peaks = [
        pd.to_numeric(r.get("Peak1_T2_ms", np.nan), errors="coerce"),
        pd.to_numeric(r.get("Peak2_T2_ms", np.nan), errors="coerce"),
        pd.to_numeric(r.get("Peak3_T2_ms", np.nan), errors="coerce"),
        pd.to_numeric(r.get("Peak4_T2_ms", np.nan), errors="coerce"),
    ]

    peaks = np.array([p for p in peaks if np.isfinite(p) and p > 0], dtype=float)

    logmean = pd.to_numeric(r.get("Weighted_logmean_T2_ms", np.nan), errors="coerce")
    width   = pd.to_numeric(r.get("Width_log10T2", np.nan), errors="coerce")
    npeak   = pd.to_numeric(r.get("N_detected_peaks", np.nan), errors="coerce")

    mobility = np.log10(logmean) if np.isfinite(logmean) and logmean > 0 else np.nan
    longtail = np.log10(np.max(peaks)) if len(peaks) > 0 else np.nan
    hetero   = width if np.isfinite(width) else np.nan
    complex_ = npeak if np.isfinite(npeak) else np.nan

    vals = np.array([mobility, longtail, hetero, complex_], dtype=float)

    if np.all(~np.isfinite(vals)):
        return np.nan

    w = np.array([0.45, 0.20, 0.25, 0.10])
    mask = np.isfinite(vals)
    w_eff = w[mask] / w[mask].sum()

    return float(np.sum(vals[mask] * w_eff))

def sample_conditioned_pool(n, monomer_list, total_active, rng):
    idx = {m: i for i, m in enumerate(monomer_list)}
    X = np.zeros((n, len(monomer_list)), dtype=float)

    hydrophilic_choices = [m for m in HYDROPHILIC if m in idx]
    hydrophobic_choices = [m for m in HYDROPHOBIC if m in idx]
    crosslink_choices   = [m for m in CROSSLINKER if m in idx]

    for i in range(n):
        frac = rng.dirichlet([1, 1, 1]) * total_active
        X[i, idx[rng.choice(hydrophilic_choices)]] = frac[0]
        X[i, idx[rng.choice(hydrophobic_choices)]] = frac[1]
        X[i, idx[rng.choice(crosslink_choices)]]   = frac[2]

    return X

def X_to_components(x, monomer_list):
    idx = {m: i for i, m in enumerate(monomer_list)}

    hphil = [(m, x[idx[m]]) for m in HYDROPHILIC if m in idx and x[idx[m]] > 0]
    hpho  = [(m, x[idx[m]]) for m in HYDROPHOBIC if m in idx and x[idx[m]] > 0]
    cross = [(m, x[idx[m]]) for m in CROSSLINKER if m in idx and x[idx[m]] > 0]

    def pick_best(lst):
        if len(lst) == 0:
            return ("Unknown", 0.0)
        return max(lst, key=lambda t: t[1])

    hp, hp_v = pick_best(hphil)
    hb, hb_v = pick_best(hpho)
    cl, cl_v = pick_best(cross)

    return hp, hb, cl, hp_v, hb_v, cl_v

def fit_candidate_feature_models(sub0, X_comp0, numeric_cols):
    models = {}
    medians = {}

    for col in numeric_cols:
        y = pd.to_numeric(sub0[col], errors="coerce").values.astype(float)
        mask = np.isfinite(y) & np.isfinite(X_comp0).all(axis=1)

        if mask.sum() >= 6 and np.nanstd(y[mask]) > 1e-12:
            model = GradientBoostingRegressor(random_state=0)
            model.fit(X_comp0[mask], y[mask])
            models[col] = model
            medians[col] = float(np.nanmedian(y[mask]))
        else:
            models[col] = None
            medians[col] = float(np.nanmedian(y[np.isfinite(y)])) if np.isfinite(y).any() else 0.0

    return models, medians

def predict_candidate_numeric_features(X_pool, models, medians, numeric_cols):
    arr = np.zeros((len(X_pool), len(numeric_cols)), dtype=float)

    for j, col in enumerate(numeric_cols):
        model = models.get(col, None)

        if model is not None:
            pred = model.predict(X_pool)
        else:
            pred = np.full(len(X_pool), medians.get(col, 0.0), dtype=float)

        arr[:, j] = np.asarray(pred, dtype=float)

    arr = np.nan_to_num(arr, nan=0.0, posinf=0.0, neginf=0.0)
    return arr

def make_candidate_caption(x, monomer_list, variant_norm, cand_numeric=None, numeric_cols=None):
    hp, hb, cl, hp_v, hb_v, cl_v = X_to_components(x, monomer_list)

    num = {}
    if cand_numeric is not None and numeric_cols is not None:
        num = {c: cand_numeric[i] for i, c in enumerate(numeric_cols)}

    parts = []

    if "s" in variant_norm:
        parts.append(
            f"Structure crosslinked copolymer architecture hydrophilic unit {hp} "
            f"hydrophobic unit {hb} crosslinker {cl} polymer network topology"
        )

    if "n" in variant_norm:
        parts.append(
            "Numeric-only quantitative composition meso-structure interaction and ILT descriptors"
        )

    if "c" in variant_norm:
        parts.append(
            "Chemistry hydration polarity hydrophobicity ionic character aromaticity "
            "fluorination hydrogen bonding network formation"
        )

    if "m" in variant_norm:
        scores = [
            num.get("headgroup_score_norm", np.nan),
            num.get("glycerol_score_norm", np.nan),
            num.get("alkenyl_score_norm", np.nan),
            num.get("alkyl_chain_score_norm", np.nan),
        ]
        motifs = ["headgroup", "glycerol", "alkenyl", "alkyl-chain"]

        if np.isfinite(scores).any():
            dom = motifs[int(np.nanargmax(scores))]
            parts.append(
                f"Meso-structure lipid-associated dominant meso-structural region {dom} "
                "headgroup glycerol alkenyl alkyl-chain association"
            )
        else:
            parts.append(
                "Meso-structure lipid-associated headgroup glycerol alkenyl alkyl-chain association"
            )

    if "i" in variant_norm:
        selectivity = num.get("motif_selectivity_index", np.nan)
        breadth = num.get("motif_breadth_count_0p5", np.nan)

        if np.isfinite(selectivity) and selectivity >= 0.35:
            sel = "highly meso-structure-selective interaction"
        elif np.isfinite(selectivity) and selectivity >= 0.18:
            sel = "moderately meso-structure-selective interaction"
        elif np.isfinite(selectivity):
            sel = "weakly selective distributed interaction"
        else:
            sel = "meso-structure-related interaction"

        if np.isfinite(breadth) and breadth >= 3:
            br = "broad multi-region perturbation"
        elif np.isfinite(breadth) and breadth >= 2:
            br = "dual-region perturbation"
        elif np.isfinite(breadth):
            br = "localized single-region perturbation"
        else:
            br = "interfacial perturbation"

        parts.append(
            f"Interaction {sel} {br} polymer-lipid localized broad perturbation "
            "meso-structure-dependent interfacial response"
        )

    if "d" in variant_norm:
        wl = num.get("Weighted_logmean_T2_ms", np.nan)
        width = num.get("Width_log10T2", np.nan)
        npeaks = num.get("N_detected_peaks", np.nan)

        dwords = [
            "Dynamics",
            "ILT-derived",
            "mobility",
            "heterogeneity",
            "relaxation",
            "landscape"
        ]

        if np.isfinite(wl):
            if wl >= 100:
                dwords.append("mobile segmental dynamics")
            elif wl >= 10:
                dwords.append("intermediate segmental mobility")
            else:
                dwords.append("constrained segmental dynamics")

        if np.isfinite(width):
            if width >= 1.2:
                dwords.append("heterogeneous broad relaxation distribution")
            elif width >= 0.6:
                dwords.append("moderately heterogeneous relaxation distribution")
            else:
                dwords.append("narrow relaxation distribution")

        if np.isfinite(npeaks):
            if npeaks >= 3:
                dwords.append("multi-component relaxation landscape")
            elif npeaks >= 2:
                dwords.append("two-component relaxation landscape")
            else:
                dwords.append("single dominant relaxation component")

        parts.append(" ".join(dwords))

    if "p" in variant_norm:
        parts.append(
            "Process thermally initiated radical polymerization polar aprotic solvent DMSO "
            "AIBN initiator curing crosslinked polymer network preparation"
        )

    if len(parts) == 0:
        parts.append(f"polymer candidate {hp}_{hb}_{cl}")

    return ". ".join(parts)

def token_from_col(col):
    if col.startswith("TOK_"):
        return col[4:]
    if col.startswith("SP_"):
        return col[3:]
    return col

def normalize_token_text(s):
    s = str(s).lower()
    s = s.replace("▁", " ")
    s = re.sub(r"[^a-z0-9_+\-\.]+", " ", s)
    return s

def captions_to_token_features(captions, feature_cols):
    tokens = [token_from_col(c) for c in feature_cols]
    token_norm = [normalize_token_text(t).strip() for t in tokens]

    X = np.zeros((len(captions), len(feature_cols)), dtype=float)

    for i, cap in enumerate(captions):
        cap_norm = normalize_token_text(cap)

        for j, tok in enumerate(token_norm):
            if tok == "":
                continue

            if " " in tok:
                X[i, j] = cap_norm.count(tok)
            else:
                pattern = r"(?<![a-z0-9_])" + re.escape(tok) + r"(?![a-z0-9_])"
                X[i, j] = len(re.findall(pattern, cap_norm))

    return X

def observed_text_features(df_sub, feature_cols):
    X = df_sub[feature_cols].copy()

    for c in feature_cols:
        X[c] = pd.to_numeric(X[c], errors="coerce")

    return X.fillna(0.0).values.astype(float)

def observed_numeric_features(df_sub, numeric_cols):
    X = df_sub[numeric_cols].copy()

    for c in numeric_cols:
        X[c] = pd.to_numeric(X[c], errors="coerce")

    X = X.replace([np.inf, -np.inf], np.nan)
    med = X.median(axis=0, numeric_only=True).fillna(0.0)
    X = X.fillna(med).fillna(0.0)

    return X.values.astype(float)

def safe_standardize(X_obs, X_cand):
    scaler = StandardScaler()
    X_obs_s = scaler.fit_transform(X_obs)
    X_cand_s = scaler.transform(X_cand)

    X_obs_s = np.nan_to_num(X_obs_s, nan=0.0, posinf=0.0, neginf=0.0)
    X_cand_s = np.nan_to_num(X_cand_s, nan=0.0, posinf=0.0, neginf=0.0)

    return X_obs_s, X_cand_s

def reduce_joint_pca(X_obs, X_cand, max_dim=32):
    X_all = np.vstack([X_obs, X_cand])

    if X_all.shape[0] <= 2:
        return X_obs, X_cand

    ncomp = min(max_dim, X_all.shape[0] - 1, X_all.shape[1])
    ncomp = max(ncomp, 1)

    pca = PCA(n_components=ncomp, random_state=0)
    Z = pca.fit_transform(X_all)

    return Z[:len(X_obs)], Z[len(X_obs):]

def normalize_pairwise_distance_scale(X_obs, X_cand):
    if len(X_obs) < 2:
        return X_obs, X_cand

    D = cdist(X_obs, X_obs)
    vals = D[np.triu_indices_from(D, k=1)]
    med = np.nanmedian(vals)

    if not np.isfinite(med) or med <= 1e-12:
        return X_obs, X_cand

    return X_obs / med, X_cand / med

def make_hybrid_repr(X_num_obs_s, X_num_cand_s, X_text_obs_pca, X_text_cand_pca):
    X_hybrid_obs = np.hstack([
        NUMERIC_WEIGHT * X_num_obs_s,
        TEXT_WEIGHT * X_text_obs_pca
    ])

    X_hybrid_cand = np.hstack([
        NUMERIC_WEIGHT * X_num_cand_s,
        TEXT_WEIGHT * X_text_cand_pca
    ])

    X_hybrid_obs, X_hybrid_cand = normalize_pairwise_distance_scale(
        X_hybrid_obs,
        X_hybrid_cand
    )

    return X_hybrid_obs, X_hybrid_cand

def occupied_fraction(P, bins=GRID_BINS, xlim=(0, 1), ylim=(0, 1)):
    if P is None or len(P) == 0:
        return 0.0

    bx, by = bins
    x = np.clip(P[:, 0], *xlim)
    y = np.clip(P[:, 1], *ylim)

    ix = np.floor((x - xlim[0]) / (xlim[1] - xlim[0] + 1e-12) * bx).astype(int)
    iy = np.floor((y - ylim[0]) / (ylim[1] - ylim[0] + 1e-12) * by).astype(int)

    ix = np.clip(ix, 0, bx - 1)
    iy = np.clip(iy, 0, by - 1)

    return len(set(zip(ix, iy))) / (bx * by)

def convex_hull_area(P):
    if P is None or len(P) < 3:
        return 0.0

    P2 = np.unique(np.asarray(P, float), axis=0)

    if len(P2) < 3:
        return 0.0

    try:
        hull = ConvexHull(P2)
        return float(hull.volume)
    except Exception:
        return 0.0

def get_mesostructure_for_selected(X_sel, numeric_models, numeric_medians, numeric_cols):
    X_sel_num_raw = predict_candidate_numeric_features(
        X_sel,
        numeric_models,
        numeric_medians,
        numeric_cols
    )

    if "motif_selectivity_index" in numeric_cols:
        idx = numeric_cols.index("motif_selectivity_index")
        meso = X_sel_num_raw[:, idx]
    else:
        meso = np.zeros(len(X_sel))

    meso = np.asarray(meso, dtype=float)
    meso = np.nan_to_num(meso, nan=0.0, posinf=0.0, neginf=0.0)

    return meso

def update_available_by_min_distance(available, X_cand_comp, pick, min_dist):
    if min_dist is None or min_dist <= 0:
        available[pick] = False
        return available

    d = np.linalg.norm(X_cand_comp - X_cand_comp[pick], axis=1)
    available = available & (d > min_dist)
    available[pick] = False
    return available

# ===============================================================
# 4. BO functions
# ===============================================================

def is_non_dominated(Y):
    Y = np.asarray(Y, dtype=float)
    n = Y.shape[0]
    nd = np.ones(n, dtype=bool)

    for i in range(n):
        if not nd[i]:
            continue

        dominates_i = (
            np.all(Y >= Y[i], axis=1)
            & np.any(Y > Y[i], axis=1)
        )

        if np.any(dominates_i):
            nd[i] = False

    return nd

def bo_acquisition(mu_dyn, sd_dyn, mu_meso, sd_meso):
    mu_dyn = np.asarray(mu_dyn, dtype=float)
    mu_meso = np.asarray(mu_meso, dtype=float)
    sd_dyn = np.asarray(sd_dyn, dtype=float)
    sd_meso = np.asarray(sd_meso, dtype=float)

    Y_mu = np.c_[mu_dyn, mu_meso]
    nd_mask = is_non_dominated(Y_mu)

    dyn_s = minmax01(mu_dyn)
    meso_s = minmax01(mu_meso)

    uncertainty_s = minmax01(sd_dyn) + minmax01(sd_meso)
    uncertainty_s = minmax01(uncertainty_s)

    balanced_score = np.sqrt(np.clip(dyn_s, 0, 1) * np.clip(meso_s, 0, 1))
    score = balanced_score + 0.15 * uncertainty_s
    score[~nd_mask] *= 0.25

    return score, nd_mask

def run_bo(
    X_obs_repr,
    X_cand_repr,
    X_cand_comp,
    y_dyn_obs,
    y_meso_obs,
    n_iter,
    init_idx
):
    Xo = X_obs_repr[init_idx].copy()
    yd = y_dyn_obs[init_idx].copy()
    ym = y_meso_obs[init_idx].copy()

    picked = []
    picked_mu_dyn = []
    picked_mu_meso = []
    picked_is_front = []

    available = np.ones(len(X_cand_repr), dtype=bool)

    for _ in range(n_iter):
        cand_idx = np.where(available)[0]

        if len(cand_idx) == 0:
            break

        kernel_d = ConstantKernel(1.0) * Matern(nu=2.5) + WhiteKernel(noise_level=1e-5)
        kernel_m = ConstantKernel(1.0) * Matern(nu=2.5) + WhiteKernel(noise_level=1e-5)

        gp_dyn = GaussianProcessRegressor(
            kernel=kernel_d,
            normalize_y=True,
            random_state=0
        )

        gp_meso = GaussianProcessRegressor(
            kernel=kernel_m,
            normalize_y=True,
            random_state=1
        )

        gp_dyn.fit(Xo, yd)
        gp_meso.fit(Xo, ym)

        mu_dyn, sd_dyn = gp_dyn.predict(X_cand_repr[cand_idx], return_std=True)
        mu_meso, sd_meso = gp_meso.predict(X_cand_repr[cand_idx], return_std=True)

        mu_dyn = np.clip(mu_dyn, 0, 1)
        mu_meso = np.clip(mu_meso, 0, 1)

        score, nd_mask = bo_acquisition(
            mu_dyn,
            sd_dyn,
            mu_meso,
            sd_meso
        )

        local_pick = int(np.argmax(score))
        pick = cand_idx[local_pick]

        picked.append(X_cand_comp[pick].copy())
        picked_mu_dyn.append(float(mu_dyn[local_pick]))
        picked_mu_meso.append(float(mu_meso[local_pick]))
        picked_is_front.append(bool(nd_mask[local_pick]))

        Xo = np.vstack([Xo, X_cand_repr[pick]])
        yd = np.append(yd, mu_dyn[local_pick])
        ym = np.append(ym, mu_meso[local_pick])

        available = update_available_by_min_distance(
            available,
            X_cand_comp,
            pick,
            MIN_COMPOSITION_DISTANCE
        )

    info = pd.DataFrame({
        "pred_mu_dynamics": picked_mu_dyn,
        "pred_mu_mesostructure": picked_mu_meso,
        "selected_from_predicted_BO_front": picked_is_front,
    })

    return np.array(picked, dtype=float), info

def run_blox(X_obs_repr, X_cand_repr, X_cand_comp, n_iter, init_idx, rng):
    observed_repr = X_obs_repr[init_idx].copy()
    picked = []
    available = np.ones(len(X_cand_repr), dtype=bool)

    for _ in range(n_iter):
        cand_available = np.where(available)[0]

        if len(cand_available) == 0:
            break

        if len(cand_available) > N_CAND:
            cand_idx = rng.choice(cand_available, size=N_CAND, replace=False)
        else:
            cand_idx = cand_available

        D = cdist(X_cand_repr[cand_idx], observed_repr)
        novelty = np.sum(np.exp(-D), axis=1)

        local_pick = int(np.argmin(novelty))
        pick = cand_idx[local_pick]

        picked.append(X_cand_comp[pick].copy())

        observed_repr = np.vstack([observed_repr, X_cand_repr[pick]])

        available = update_available_by_min_distance(
            available,
            X_cand_comp,
            pick,
            MIN_COMPOSITION_DISTANCE
        )

    return np.array(picked, dtype=float)

def add_title(slide, title, subtitle=None):
    title_box = slide.shapes.add_textbox(
        Inches(0.45), Inches(0.25), Inches(12.4), Inches(0.45)
    )
    tf = title_box.text_frame
    tf.clear()
    p = tf.paragraphs[0]
    p.text = title
    p.font.size = Pt(22)
    p.font.bold = True

    if subtitle is not None:
        sub_box = slide.shapes.add_textbox(
            Inches(0.45), Inches(0.72), Inches(12.4), Inches(0.30)
        )
        tf = sub_box.text_frame
        tf.clear()
        p = tf.paragraphs[0]
        p.text = subtitle
        p.font.size = Pt(11)

def add_image_fit(slide, image_path, left, top, width, height):
    if os.path.exists(image_path):
        slide.shapes.add_picture(
            image_path,
            Inches(left),
            Inches(top),
            width=Inches(width),
            height=Inches(height)
        )

# ===============================================================
# 5. Preprocess
# ===============================================================

df["Copolymer_Name_norm"] = df["Copolymer_Name"].apply(normalize_name)
df["variant_norm"] = df["variant"].apply(normalize_variant)

df = df[df["variant_norm"].isin(TARGET_VARIANTS)].copy().reset_index(drop=True)

if len(df) == 0:
    raise ValueError("No rows found for TARGET_VARIANTS.")

df["Dynamics_raw_local"] = df.apply(compute_ilt_dynamics_score_row, axis=1)

source_df = df[df["variant_norm"] == DYNAMICS_SOURCE_VARIANT].copy()

if len(source_df) == 0 or source_df["Dynamics_raw_local"].notna().sum() == 0:
    print(f"[WARN] No valid dynamics for {DYNAMICS_SOURCE_VARIANT}; using all variants.")
    source_df = df.copy()

source_dyn = (
    source_df[["Copolymer_Name", "Dynamics_raw_local"]]
    .dropna(subset=["Dynamics_raw_local"])
    .groupby("Copolymer_Name", as_index=False)["Dynamics_raw_local"]
    .mean()
    .rename(columns={"Dynamics_raw_local": "Dynamics_raw_shared"})
)

df = df.merge(source_dyn, on="Copolymer_Name", how="left")

finite = np.isfinite(df["Dynamics_raw_shared"].values.astype(float))
df["Dynamics"] = np.nan
df.loc[finite, "Dynamics"] = scale01(df.loc[finite, "Dynamics_raw_shared"].values)

df["Hydrophobic_raw"] = df.apply(
    lambda r: hydrophobic_fraction_from_name_and_comps(
        r["Copolymer_Name_norm"],
        r.get("comp_1", np.nan),
        r.get("comp_2", np.nan),
        r.get("comp_3", np.nan),
        TOTAL_ACTIVE_DEFAULT
    ),
    axis=1
)

hydro = df["Hydrophobic_raw"].values.astype(float)
hydro_fill = np.nanmedian(hydro[np.isfinite(hydro)]) if np.isfinite(hydro).any() else 0.0

df["Hydrophobic"] = np.where(np.isfinite(hydro), hydro, hydro_fill)
df["Hydrophobic"] = np.clip(df["Hydrophobic"], 0, 1)

variants = [v for v in TARGET_VARIANTS if v in sorted(df["variant_norm"].unique())]

print("Variants used:", variants)
print("Variant labels:", [VARIANT_LABEL_MAP.get(v, v.upper()) for v in variants])
print("Rows used:", df.shape)

# ===============================================================
# 6. Main loop
# ===============================================================

SUMMARY_ROWS = []
POINT_ROWS = []
REP_DIAGNOSTIC_ROWS = []
BO_INFO_ROWS = []

for variant_norm in variants:

    variant_label = VARIANT_LABEL_MAP.get(variant_norm, variant_norm.upper())

    print(f"\n=== Variant: {variant_label} ===")

    sub0 = df[df["variant_norm"] == variant_norm].copy().reset_index(drop=True)

    X_comp0 = build_observed_composition_matrix(sub0, ALL_MONOMERS)

    valid_comp = X_comp0.sum(axis=1) > 0
    sub0 = sub0.loc[valid_comp].reset_index(drop=True)
    X_comp0 = X_comp0[valid_comp]

    dyn0 = sub0["Dynamics"].values.astype(float)

    meso0_raw = pd.to_numeric(
        sub0["motif_selectivity_index"],
        errors="coerce"
    ).values.astype(float)

    if np.isfinite(meso0_raw).any():
        meso_fill = np.nanmedian(meso0_raw[np.isfinite(meso0_raw)])
        meso0_filled = np.where(np.isfinite(meso0_raw), meso0_raw, meso_fill)
        meso0 = scale01(meso0_filled)
    else:
        meso0 = np.zeros_like(dyn0)

    valid = (
        np.isfinite(dyn0)
        & np.isfinite(meso0)
        & np.isfinite(X_comp0).all(axis=1)
    )

    sub0 = sub0.loc[valid].reset_index(drop=True)
    X_comp0 = X_comp0[valid]
    dyn0 = dyn0[valid]
    meso0 = meso0[valid]

    if len(sub0) < 8:
        print("Skipped due to too few valid samples.")
        continue

    X_text_obs_raw0 = observed_text_features(sub0, TEXT_FEATURE_COLS)
    X_num_obs_blox_raw0 = observed_numeric_features(sub0, NUMERIC_REPR_COLS_BLOX)
    X_num_obs_bo_raw0 = observed_numeric_features(sub0, NUMERIC_REPR_COLS_BO)

    total_active_obs = np.nanmedian(
        sub0[["comp_1", "comp_2", "comp_3"]].sum(axis=1).values.astype(float)
    )

    if not np.isfinite(total_active_obs) or total_active_obs <= 0:
        total_active_obs = TOTAL_ACTIVE_DEFAULT

    dyn_model = GradientBoostingRegressor(random_state=0)
    dyn_model.fit(X_comp0, dyn0)

    def pred_dyn(X):
        return np.clip(dyn_model.predict(np.asarray(X)), 0, 1)

    numeric_models_blox, numeric_medians_blox = fit_candidate_feature_models(
        sub0,
        X_comp0,
        NUMERIC_REPR_COLS_BLOX
    )

    numeric_models_bo, numeric_medians_bo = fit_candidate_feature_models(
        sub0,
        X_comp0,
        NUMERIC_REPR_COLS_BO
    )

    for repeat in range(N_REPEATS):

        rng = np.random.default_rng(1000 + repeat)

        POOL = sample_conditioned_pool(
            N_POOL,
            ALL_MONOMERS,
            total_active_obs,
            rng
        )

        X_num_cand_blox_raw = predict_candidate_numeric_features(
            POOL,
            numeric_models_blox,
            numeric_medians_blox,
            NUMERIC_REPR_COLS_BLOX
        )

        X_num_cand_bo_raw = predict_candidate_numeric_features(
            POOL,
            numeric_models_bo,
            numeric_medians_bo,
            NUMERIC_REPR_COLS_BO
        )

        cand_captions_blox = [
            make_candidate_caption(
                x,
                ALL_MONOMERS,
                variant_norm,
                cand_numeric=X_num_cand_blox_raw[i],
                numeric_cols=NUMERIC_REPR_COLS_BLOX
            )
            for i, x in enumerate(POOL)
        ]

        cand_captions_bo = [
            make_candidate_caption(
                x,
                ALL_MONOMERS,
                variant_norm,
                cand_numeric=X_num_cand_bo_raw[i],
                numeric_cols=NUMERIC_REPR_COLS_BO
            )
            for i, x in enumerate(POOL)
        ]

        X_text_cand_blox_raw = captions_to_token_features(
            cand_captions_blox,
            TEXT_FEATURE_COLS
        )

        X_text_cand_bo_raw = captions_to_token_features(
            cand_captions_bo,
            TEXT_FEATURE_COLS
        )

        nonzero_fraction = float(
            np.mean(np.sum(np.abs(X_text_cand_blox_raw), axis=1) > 0)
        )

        mean_nonzero_features = float(
            np.mean(np.sum(np.abs(X_text_cand_blox_raw) > 0, axis=1))
        )

        X_num_obs_blox_s, X_num_cand_blox_s = safe_standardize(
            X_num_obs_blox_raw0,
            X_num_cand_blox_raw
        )

        X_num_blox_obs, X_num_blox_cand = normalize_pairwise_distance_scale(
            X_num_obs_blox_s,
            X_num_cand_blox_s
        )

        X_text_obs_blox_s, X_text_cand_blox_s = safe_standardize(
            X_text_obs_raw0,
            X_text_cand_blox_raw
        )

        X_text_blox_obs_pca, X_text_blox_cand_pca = reduce_joint_pca(
            X_text_obs_blox_s,
            X_text_cand_blox_s,
            max_dim=32
        )

        X_hybrid_blox_obs, X_hybrid_blox_cand = make_hybrid_repr(
            X_num_obs_blox_s,
            X_num_cand_blox_s,
            X_text_blox_obs_pca,
            X_text_blox_cand_pca
        )

        X_num_obs_bo_s, X_num_cand_bo_s = safe_standardize(
            X_num_obs_bo_raw0,
            X_num_cand_bo_raw
        )

        X_num_bo_obs, X_num_bo_cand = normalize_pairwise_distance_scale(
            X_num_obs_bo_s,
            X_num_cand_bo_s
        )

        X_text_obs_bo_s, X_text_cand_bo_s = safe_standardize(
            X_text_obs_raw0,
            X_text_cand_bo_raw
        )

        X_text_bo_obs_pca, X_text_bo_cand_pca = reduce_joint_pca(
            X_text_obs_bo_s,
            X_text_cand_bo_s,
            max_dim=32
        )

        X_hybrid_bo_obs, X_hybrid_bo_cand = make_hybrid_repr(
            X_num_obs_bo_s,
            X_num_cand_bo_s,
            X_text_bo_obs_pca,
            X_text_bo_cand_pca
        )

        REP_DIAGNOSTIC_ROWS.append({
            "variant": variant_norm,
            "variant_label": variant_label,
            "repeat": repeat,
            "candidate_caption_nonzero_fraction": nonzero_fraction,
            "candidate_text_mean_nonzero_features": mean_nonzero_features,
            "numeric_repr_dim_blox": X_num_obs_blox_raw0.shape[1],
            "numeric_repr_dim_bo_leakfree": X_num_obs_bo_raw0.shape[1],
            "text_repr_dim": X_text_obs_raw0.shape[1],
            "hybrid_repr_dim_blox": X_hybrid_blox_obs.shape[1],
            "hybrid_repr_dim_bo_leakfree": X_hybrid_bo_obs.shape[1],
            "bo_removed_leakage_cols": ";".join(BO_LEAKAGE_COLS),
            "numeric_weight": NUMERIC_WEIGHT,
            "text_weight": TEXT_WEIGHT,
            "min_composition_distance": MIN_COMPOSITION_DISTANCE,
            "bo_type": "bo_leakfree_numeric",
        })

        init_n = min(5, len(dyn0))
        init_idx = rng.choice(len(dyn0), size=init_n, replace=False)

        selected = {}
        bo_infos = {}

        selected["BO-Numerical"], bo_infos["BO-Numerical"] = run_bo(
            X_num_bo_obs,
            X_num_bo_cand,
            POOL,
            dyn0,
            meso0,
            BO_ITERS,
            init_idx
        )

        selected["BO-N+Languange"], bo_infos["BO-N+Languange"] = run_bo(
            X_hybrid_bo_obs,
            X_hybrid_bo_cand,
            POOL,
            dyn0,
            meso0,
            BO_ITERS,
            init_idx
        )

        selected["BLOX-Numerical"] = run_blox(
            X_num_blox_obs,
            X_num_blox_cand,
            POOL,
            BLOX_ITERS,
            init_idx,
            rng
        )

        selected["BLOX-N+Languange"] = run_blox(
            X_hybrid_blox_obs,
            X_hybrid_blox_cand,
            POOL,
            BLOX_ITERS,
            init_idx,
            rng
        )

        for method, X_sel in selected.items():

            if len(X_sel) == 0:
                continue

            dyn_x = pred_dyn(X_sel)
            hydro_y = hydrophobic_from_X(X_sel, ALL_MONOMERS, total_active_obs)

            meso_y_raw = get_mesostructure_for_selected(
                X_sel,
                numeric_models_blox,
                numeric_medians_blox,
                NUMERIC_REPR_COLS_BLOX
            )

            meso_y_scaled = scale01(meso_y_raw)

            print(
                f"{variant_label} | repeat {repeat} | {method} | "
                f"Dynamics: {np.nanmin(dyn_x):.3f}-{np.nanmax(dyn_x):.3f} | "
                f"Meso raw: {np.nanmin(meso_y_raw):.3f}-{np.nanmax(meso_y_raw):.3f}"
            )

            P_dh = np.c_[
                np.clip(dyn_x, 0, 1),
                np.clip(hydro_y, 0, 1)
            ]

            P_dm = np.c_[
                np.clip(dyn_x, 0, 1),
                np.clip(meso_y_scaled, 0, 1)
            ]

            alg = "BO" if method.startswith("BO") else "BLOX"
            rep = "N+Languange" if "Languange" in method else "Numerical"

            SUMMARY_ROWS.append({
                "variant": variant_norm,
                "variant_label": variant_label,
                "repeat": repeat,
                "method": method,
                "algorithm": alg,
                "representation": rep,
                "n_points": len(P_dh),

                "occupied_fraction_Dynamics_Hydrophobic": occupied_fraction(P_dh),
                "occupied_fraction_Dynamics_Mesostructure": occupied_fraction(P_dm),

                "convex_hull_area_Dynamics_Hydrophobic": convex_hull_area(P_dh),
                "convex_hull_area_Dynamics_Mesostructure": convex_hull_area(P_dm),

                "mean_dynamics": float(np.mean(P_dh[:, 0])),
                "mean_hydrophobicity": float(np.mean(P_dh[:, 1])),
                "mean_mesostructure_scaled": float(np.mean(P_dm[:, 1])),
                "mean_mesostructure_raw": float(np.mean(meso_y_raw)),

                "range_dynamics": float(np.max(P_dh[:, 0]) - np.min(P_dh[:, 0])),
                "range_hydrophobicity": float(np.max(P_dh[:, 1]) - np.min(P_dh[:, 1])),
                "range_mesostructure_scaled": float(np.max(P_dm[:, 1]) - np.min(P_dm[:, 1])),
                "range_mesostructure_raw": float(np.max(meso_y_raw) - np.min(meso_y_raw)),

                "best_dynamics": float(np.max(P_dh[:, 0])),
                "best_mesostructure_scaled": float(np.max(P_dm[:, 1])),
                "best_mesostructure_raw": float(np.max(meso_y_raw)),
            })

            for i, (dx, hy, my_s, my_raw) in enumerate(
                zip(P_dh[:, 0], P_dh[:, 1], P_dm[:, 1], meso_y_raw)
            ):
                POINT_ROWS.append({
                    "variant": variant_norm,
                    "variant_label": variant_label,
                    "repeat": repeat,
                    "method": method,
                    "algorithm": alg,
                    "representation": rep,
                    "point_id": i + 1,
                    "Dynamics_x": float(dx),
                    "Hydrophobicity_y": float(hy),
                    "Mesostructure_y_scaled": float(my_s),
                    "Mesostructure_y_raw": float(my_raw),
                })

            if method in bo_infos:
                pinfo = bo_infos[method].copy()
                pinfo["variant"] = variant_norm
                pinfo["variant_label"] = variant_label
                pinfo["repeat"] = repeat
                pinfo["method"] = method
                pinfo["point_id"] = np.arange(1, len(pinfo) + 1)
                BO_INFO_ROWS.extend(pinfo.to_dict("records"))

    # -----------------------------------------------------------
    # Plot all repeats for each variant
    # -----------------------------------------------------------

    points_now = pd.DataFrame(POINT_ROWS)
    points_now = points_now[points_now["variant"] == variant_norm].copy()

    if len(points_now) > 0:

        # Dynamics × Hydrophobicity
        fig, ax = plt.subplots(figsize=(7.8, 5.8))

        for method in METHODS:
            p = points_now[points_now["method"] == method]
            if len(p) == 0:
                continue

            ax.scatter(
                p["Dynamics_x"],
                p["Hydrophobicity_y"],
                c=METHOD_COLORS[method],
                marker=METHOD_MARKERS[method],
                s=35,
                alpha=0.35,
                edgecolor="black",
                linewidth=0.25,
                label=method
            )

        ax.set_xlabel("Dynamics", fontsize=LABEL_FS)
        ax.set_ylabel("Hydrophobic fraction", fontsize=LABEL_FS)
        ax.set_title(
            f"{variant_label} | Dynamics vs Hydrophobicity",
            fontsize=TITLE_FS
        )
        ax.tick_params(labelsize=TICK_FS)
        ax.set_xlim(-0.02, 1.02)
        ax.set_ylim(-0.02, 1.02)

        ax.set_xticks([0, 0.5, 1.0])
        ax.set_yticks([0, 0.5, 1.0])

        ax.legend(
            fontsize=LEGEND_FS,
            frameon=True,
            loc="upper left",
            bbox_to_anchor=(1.02, 1.0)
        )
        plt.tight_layout()

        plt.savefig(
            os.path.join(OUTDIR, f"2D_allrepeats_Dynamics_vs_Hydrophobicity_{variant_norm}.png"),
            dpi=300,
            bbox_inches="tight"
        )
        plt.savefig(
            os.path.join(OUTDIR, f"2D_allrepeats_Dynamics_vs_Hydrophobicity_{variant_norm}.pdf"),
            bbox_inches="tight"
        )
        plt.show()
        plt.close()

        # Dynamics × Meso-structure
        fig, ax = plt.subplots(figsize=(7.8, 5.8))

        for method in METHODS:
            p = points_now[points_now["method"] == method]
            if len(p) == 0:
                continue

            ax.scatter(
                p["Dynamics_x"],
                p["Mesostructure_y_scaled"],
                c=METHOD_COLORS[method],
                marker=METHOD_MARKERS[method],
                s=35,
                alpha=0.35,
                edgecolor="black",
                linewidth=0.25,
                label=method
            )

        ax.set_xlabel("Dynamics", fontsize=LABEL_FS)
        ax.set_ylabel("Meso-structure", fontsize=LABEL_FS)
        ax.set_title(
            f"{variant_label} | Dynamics vs Meso-structure",
            fontsize=TITLE_FS
        )
        ax.tick_params(labelsize=TICK_FS)
        ax.set_xlim(-0.02, 1.02)
        ax.set_ylim(-0.02, 1.02)

        ax.set_xticks([0, 0.5, 1.0])
        ax.set_yticks([0, 0.5, 1.0])

        ax.legend(
            fontsize=LEGEND_FS,
            frameon=True,
            loc="upper left",
            bbox_to_anchor=(1.02, 1.0)
        )
        plt.tight_layout()

        plt.savefig(
            os.path.join(OUTDIR, f"2D_allrepeats_Dynamics_vs_Mesostructure_{variant_norm}.png"),
            dpi=300,
            bbox_inches="tight"
        )
        plt.savefig(
            os.path.join(OUTDIR, f"2D_allrepeats_Dynamics_vs_Mesostructure_{variant_norm}.pdf"),
            bbox_inches="tight"
        )
        plt.show()
        plt.close()

# ===============================================================
# 7. Save tables
# ===============================================================

summary_df = pd.DataFrame(SUMMARY_ROWS)
points_df = pd.DataFrame(POINT_ROWS)
rep_diag_df = pd.DataFrame(REP_DIAGNOSTIC_ROWS)
bo_info_df = pd.DataFrame(BO_INFO_ROWS)

summary_df.to_csv(
    os.path.join(OUTDIR, "summary_BO_DynamicsHydrophobicity_and_DynamicsMesostructure.csv"),
    index=False,
    encoding="utf-8-sig"
)

points_df.to_csv(
    os.path.join(OUTDIR, "points_allrepeats_BO_DynamicsHydrophobicity_and_DynamicsMesostructure.csv"),
    index=False,
    encoding="utf-8-sig"
)

rep_diag_df.to_csv(
    os.path.join(OUTDIR, "representation_diagnostics.csv"),
    index=False,
    encoding="utf-8-sig"
)

bo_info_df.to_csv(
    os.path.join(OUTDIR, "bo_selection_diagnostics.csv"),
    index=False,
    encoding="utf-8-sig"
)

# ===============================================================
# 8. Summary plots
# ===============================================================

if len(summary_df) > 0:

    for metric, title, fname in [
        (
            "occupied_fraction_Dynamics_Hydrophobic",
            "Occupied fraction: Dynamics × Hydrophobicity",
            "OF_Dynamics_Hydrophobicity_by_variant"
        ),
        (
            "occupied_fraction_Dynamics_Mesostructure",
            "Occupied fraction: Dynamics × Meso-structure",
            "OF_Dynamics_Mesostructure_by_variant"
        ),
    ]:
        mean_df = (
            summary_df
            .groupby(["variant", "variant_label", "method"], as_index=False)
            .agg(mean=(metric, "mean"), std=(metric, "std"))
        )

        order_df = pd.DataFrame({
            "variant": variants,
            "variant_label": [VARIANT_LABEL_MAP.get(v, v.upper()) for v in variants]
        })

        mean_df = mean_df.merge(order_df, on=["variant", "variant_label"], how="right")

        pivot_mean = mean_df.pivot(index="variant_label", columns="method", values="mean")
        pivot_std = mean_df.pivot(index="variant_label", columns="method", values="std")

        ordered_labels = [VARIANT_LABEL_MAP.get(v, v.upper()) for v in variants]
        pivot_mean = pivot_mean.reindex(ordered_labels)
        pivot_std = pivot_std.reindex(ordered_labels)

        fig, ax = plt.subplots(figsize=(11.5, 5.8))

        x = np.arange(len(pivot_mean.index))
        width = 0.19

        for i, method in enumerate(METHODS):
            if method not in pivot_mean.columns:
                continue

            vals = pivot_mean[method].values
            errs = pivot_std[method].values

            ax.bar(
                x + (i - 1.5) * width,
                vals,
                width,
                yerr=errs,
                capsize=3,
                label=method,
                color=METHOD_COLORS[method],
                edgecolor="black",
                linewidth=0.35
            )

        ax.set_xticks(x)
        ax.set_xticklabels(
            list(pivot_mean.index),
            fontsize=TICK_FS,
            rotation=30,
            ha="right"
        )

        ax.set_ylabel("Occupied fraction", fontsize=LABEL_FS)
        ax.tick_params(axis="y", labelsize=TICK_FS)
        ax.set_title(title, fontsize=TITLE_FS)

        ax.set_ylim(0, 0.4)
        ax.set_yticks([0, 0.2, 0.4])

        ax.legend(fontsize=10, frameon=True, ncol=2)

        plt.tight_layout()

        plt.savefig(os.path.join(OUTDIR, fname + ".png"), dpi=300, bbox_inches="tight")
        plt.savefig(os.path.join(OUTDIR, fname + ".pdf"), bbox_inches="tight")
        plt.show()
        plt.close()

    agg = (
        summary_df
        .groupby("method", as_index=False)
        .agg(
            OF_DH_mean=("occupied_fraction_Dynamics_Hydrophobic", "mean"),
            OF_DH_std=("occupied_fraction_Dynamics_Hydrophobic", "std"),
            OF_DMeso_mean=("occupied_fraction_Dynamics_Mesostructure", "mean"),
            OF_DMeso_std=("occupied_fraction_Dynamics_Mesostructure", "std"),
            mean_dynamics=("mean_dynamics", "mean"),
            best_dynamics=("best_dynamics", "mean"),
            mean_mesostructure_scaled=("mean_mesostructure_scaled", "mean"),
            best_mesostructure_scaled=("best_mesostructure_scaled", "mean"),
            mean_mesostructure_raw=("mean_mesostructure_raw", "mean"),
            best_mesostructure_raw=("best_mesostructure_raw", "mean"),
        )
    )

    agg.to_csv(
        os.path.join(OUTDIR, "algorithm_representation_summary_BO_DH_DMeso.csv"),
        index=False,
        encoding="utf-8-sig"
    )

# ===============================================================
# 9. PPT output
# ===============================================================

prs = Presentation()
prs.slide_width = Inches(13.333)
prs.slide_height = Inches(7.5)
blank = prs.slide_layouts[6]

slide = prs.slides.add_slide(blank)

title_box = slide.shapes.add_textbox(Inches(0.65), Inches(2.10), Inches(12.0), Inches(0.8))
tf = title_box.text_frame
tf.clear()
p = tf.paragraphs[0]
p.text = "BO for Dynamics–Meso-structure Exploration"
p.font.size = Pt(27)
p.font.bold = True
p.alignment = PP_ALIGN.CENTER

sub_box = slide.shapes.add_textbox(Inches(0.85), Inches(3.05), Inches(11.6), Inches(0.8))
tf = sub_box.text_frame
tf.clear()
p = tf.paragraphs[0]
p.text = "BLOX uses the original representation; BO uses leak-free explanatory variables"
p.font.size = Pt(14)
p.alignment = PP_ALIGN.CENTER

summary_slides = [
    (
        "OF_Dynamics_Hydrophobicity_by_variant.png",
        "Occupied fraction in Dynamics × Hydrophobicity space",
        "Mean ± SD across repeats"
    ),
    (
        "OF_Dynamics_Mesostructure_by_variant.png",
        "Occupied fraction in Dynamics × Meso-structure space",
        "Mean ± SD across repeats"
    ),
]

for img, title, subtitle in summary_slides:
    path = os.path.join(OUTDIR, img)

    if os.path.exists(path):
        slide = prs.slides.add_slide(blank)
        add_title(slide, title, subtitle)
        add_image_fit(slide, path, 1.05, 1.15, 11.2, 5.8)

for variant_norm in variants:
    variant_label = VARIANT_LABEL_MAP.get(variant_norm, variant_norm.upper())

    for prefix, title_suffix in [
        ("2D_allrepeats_Dynamics_vs_Hydrophobicity", "Dynamics × Hydrophobicity"),
        ("2D_allrepeats_Dynamics_vs_Mesostructure", "Dynamics × Meso-structure"),
    ]:
        path = os.path.join(OUTDIR, f"{prefix}_{variant_norm}.png")

        if os.path.exists(path):
            slide = prs.slides.add_slide(blank)
            add_title(
                slide,
                f"{title_suffix} | {variant_label}",
                "All repeats shown with transparent markers"
            )
            add_image_fit(slide, path, 0.85, 1.10, 11.8, 6.0)

prs.save(PPTX_PATH)

# ===============================================================
# 10. ZIP
# ===============================================================

with zipfile.ZipFile(ZIP_PATH, "w", zipfile.ZIP_DEFLATED) as zf:
    for fn in os.listdir(OUTDIR):
        fp = os.path.join(OUTDIR, fn)

        if os.path.isfile(fp):
            zf.write(fp, arcname=fn)

    if os.path.exists(PPTX_PATH):
        zf.write(PPTX_PATH, arcname=os.path.basename(PPTX_PATH))

print("\nDone.")
print("Saved files in:", OUTDIR)
print("PPTX saved to :", PPTX_PATH)
print("ZIP saved to  :", ZIP_PATH)

print("\nKey outputs:")
print("- summary_BO_DynamicsHydrophobicity_and_DynamicsMesostructure.csv")
print("- points_allrepeats_BO_DynamicsHydrophobicity_and_DynamicsMesostructure.csv")
print("- bo_selection_diagnostics.csv")
print("- representation_diagnostics.csv")
print("- algorithm_representation_summary_BO_DH_DMeso.csv")
print("- OF_Dynamics_Hydrophobicity_by_variant.png")
print("- OF_Dynamics_Mesostructure_by_variant.png")
print("- 2D_allrepeats_Dynamics_vs_Mesostructure_<variant>.png")
print("- BO_Dynamics_Mesostructure_output.pptx")
print("- BO_Dynamics_Mesostructure_output.zip")
```
