```python
import pandas as pd
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# Color settings (GREEN palette)
# ------------------------------------------------------------
COLOR_MAP = {
    "AMPS_HMA_DICL": "#1B5E20",  # dark green
    "AMPS_HMA_TRCL": "#43A047",  # medium green
    "AMPS_HMA_TECL": "#A5D6A7"   # light green
}

CSV_CPMG = "AddD2O_CPMG_T2_Relative_Intensity_smoothed_44copolymers.csv"
samples = ["AMPS_HMA_DICL", "AMPS_HMA_TRCL", "AMPS_HMA_TECL"]

# ------------------------------------------------------------
# Load data
# ------------------------------------------------------------
df = pd.read_csv(CSV_CPMG)
df = df[df.iloc[:, 0].isin(samples)]

plt.figure(figsize=(8, 6))

# ------------------------------------------------------------
# Plot
# ------------------------------------------------------------
for _, row in df.iterrows():
    name = row.iloc[0]
    y = row.iloc[1:].values
    x = range(len(y))
    plt.plot(
        x, y,
        lw=3,
        label=name,
        color=COLOR_MAP[name]
    )

# ------------------------------------------------------------
# Axes & ticks
# ------------------------------------------------------------
plt.xlabel("Time (ms)", fontsize=22)
plt.ylabel("Relative Intensity", fontsize=22)

plt.xticks(
    [0, len(x)//2, len(x)-1],
    ["0", "400", "800"],
    fontsize=22
)

# Y-axis: exactly 3 ticks
ymin, ymax = plt.ylim()
plt.yticks(
    [ymin, (ymin + ymax) / 2, ymax],
    [f"{ymin:.3f}", f"{(ymin + ymax) / 2:.3f}", f"{ymax:.3f}"],
    fontsize=22
)

# ------------------------------------------------------------
# Legend & style
# ------------------------------------------------------------
plt.legend(frameon=False, fontsize=20)

ax = plt.gca()
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()
plt.savefig("CPMG_maintext_green_palette.png", dpi=300)
plt.show()
```
