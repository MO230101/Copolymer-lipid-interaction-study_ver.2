```python
# ============================================================
# COMPLETE CODE
# ROI-expanded quantification + Blank-relative quantification
# + overlay plots + quantified-feature HEATMAPS + PPT export
# + MODIFIED:
#   - Overlay Y-axis title = Intensity
#   - Overlay X-axis title = ¹H chemical shift (ppm)
#   - Overlay Y ticks fixed to 0, 0.5, 1.0
#   - Larger overlay font size
#   - Multi-window bottom ppm labels removed
# ============================================================

!pip -q install nmrglue pandas numpy matplotlib scipy openpyxl xlrd python-pptx xlsxwriter

import os
import re
import glob
import zipfile
import shutil
from pathlib import Path
from io import StringIO

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import nmrglue as ng

from scipy.signal import savgol_filter, medfilt, find_peaks, peak_widths
from scipy.ndimage import gaussian_filter1d
from pptx import Presentation
from pptx.util import Inches, Pt

from google.colab import files
uploaded = files.upload()

all_uploaded_files = list(uploaded.keys())
print("Uploaded files:")
for fn in all_uploaded_files:
    print(" -", fn)

def normalize_colab_name(fn):
    base = os.path.basename(fn)
    base = re.sub(r"\s*\(\d+\)(?=\.[^.]+$)", "", base)
    return base

def find_first_by_patterns(candidates, patterns):
    for pat in patterns:
        rx = re.compile(pat, flags=re.IGNORECASE)
        for fn in candidates:
            norm = normalize_colab_name(fn)
            if rx.search(norm):
                return fn
    return None

def stem_no_ext(path):
    return Path(normalize_colab_name(path)).stem

def normalize_basename_only(path):
    return normalize_colab_name(os.path.basename(path))

def roi_sort_key(x):
    m = re.search(r"(\d+)", str(x))
    return int(m.group(1)) if m else 999

zip_path = find_first_by_patterns(all_uploaded_files, [
    r"^ucsf\.zip$",
    r".*\.zip$"
])

csv_path = find_first_by_patterns(all_uploaded_files, [
    r"^sample_1_43_ucsf_mapping\.csv$",
    r".*mapping.*\.csv$"
])

blank_path = find_first_by_patterns(all_uploaded_files, [
    r"^Blank1\.ucsf$",
    r"^Blank.*\.ucsf$",
    r"^blank.*\.ucsf$"
])

roi_path = find_first_by_patterns(all_uploaded_files, [
    r"^roiTable.*zgpg30.*\.(xls|xlsx|csv|txt)$",
    r".*roi.*\.(xls|xlsx|csv|txt)$"
])

if zip_path is None:
    raise FileNotFoundError("ucsf.zip が見つかりません。")
if csv_path is None:
    raise FileNotFoundError("sample_1_43_ucsf_mapping.csv が見つかりません。")
if blank_path is None:
    raise FileNotFoundError("Blank1.ucsf が見つかりません。")
if roi_path is None:
    raise FileNotFoundError("roiTable_zgpg30_602.xls / .txt が見つかりません。")

print("\nDetected:")
print(" ZIP   :", zip_path)
print(" CSV   :", csv_path)
print(" BLANK :", blank_path)
print(" ROI   :", roi_path)

extract_dir = "extracted_ucsf"
if os.path.exists(extract_dir):
    shutil.rmtree(extract_dir)
os.makedirs(extract_dir, exist_ok=True)

with zipfile.ZipFile(zip_path, "r") as zf:
    zf.extractall(extract_dir)

all_ucsf_files = sorted(glob.glob(os.path.join(extract_dir, "**", "*.ucsf"), recursive=True))
all_ucsf_files += [fn for fn in all_uploaded_files if fn.lower().endswith(".ucsf")]
all_ucsf_files = sorted(set(all_ucsf_files))

print("\nTotal UCSF candidates:", len(all_ucsf_files))

df = pd.read_csv(csv_path)

required_cols = ["No", "Sample_Name", "Component_Number", "zgpg30_ucsf_No"]
for c in required_cols:
    if c not in df.columns:
        raise ValueError(f"Required column not found: {c}")

df = df.sort_values("No").reset_index(drop=True)
df["No"] = df["No"].astype(int)
df["Component_Number"] = (
    df["Component_Number"].astype(str).str.strip().str.replace("-", "_", regex=False)
)

print("\nMapping preview:")
display(df.head())

def read_1d_ucsf_manual(fp):
    sp = ng.sparky
    with open(fp, "rb") as f:
        dic = sp.fileheader2dic(sp.get_fileheader(f))
        if dic["naxis"] != 1:
            raise ValueError(f"Not a 1D UCSF file: {fp}")
        dic["w1"] = sp.axisheader2dic(sp.get_axisheader(f))
        raw = sp.get_data(f)

    npts = int(dic["w1"]["npoints"])
    data = np.asarray(raw[:npts], dtype=np.float64)
    uc = sp.make_uc(dic, data, dim=0)
    ppm = uc.ppm_scale()
    return dic, data, ppm

def try_read_1d(fp):
    try:
        _ = read_1d_ucsf_manual(fp)
        return True
    except Exception:
        return False

readable_1d_files = [fp for fp in all_ucsf_files if try_read_1d(fp)]
print("Readable 1D UCSF:", len(readable_1d_files))

MANUAL_FILE_MAP = {
    7: "1513.ucsf",
    23: "1633.ucsf"
}

def resolve_manual_file(no, file_list):
    if no not in MANUAL_FILE_MAP:
        return None
    target = MANUAL_FILE_MAP[no]
    for fp in file_list:
        if normalize_basename_only(fp) == target:
            return fp
    return None

def resolve_zgpg30_file(no, zgpg30_no, component_number, file_list):
    fp = resolve_manual_file(no, file_list)
    if fp is not None:
        return fp

    candidates = []
    zno = None

    if pd.notna(zgpg30_no):
        try:
            zno = str(int(float(zgpg30_no)))
        except Exception:
            zno = str(zgpg30_no).strip()

    comp = str(component_number).strip().replace("-", "_")

    if zno is not None:
        for fp in file_list:
            if stem_no_ext(fp) == zno:
                candidates.append(fp)
    if candidates:
        return candidates[0]

    if zno is not None:
        regex_num = re.compile(rf"(^|[_\- ]){re.escape(zno)}($|[_\- ])")
        for fp in file_list:
            stem = stem_no_ext(fp)
            if regex_num.search(stem):
                candidates.append(fp)
    if candidates:
        return candidates[0]

    for fp in file_list:
        stem = stem_no_ext(fp).replace("-", "_")
        if stem == comp:
            candidates.append(fp)
    if candidates:
        return candidates[0]

    for fp in file_list:
        stem = stem_no_ext(fp).replace("-", "_")
        if stem.startswith(comp):
            candidates.append(fp)
    if candidates:
        return candidates[0]

    return None

resolved_records = []
for _, row in df.iterrows():
    fp = resolve_zgpg30_file(
        int(row["No"]),
        row["zgpg30_ucsf_No"],
        row["Component_Number"],
        readable_1d_files
    )
    resolved_records.append({
        "No": int(row["No"]),
        "Sample_Name": row["Sample_Name"],
        "Component_Number": row["Component_Number"],
        "zgpg30_ucsf_No": row["zgpg30_ucsf_No"],
        "Resolved_File": fp if fp is not None else ""
    })

resolved_df = pd.DataFrame(resolved_records)
resolved_df.to_csv("resolved_zgpg30_1d.csv", index=False)

print("\nResolved file table:")
display(resolved_df.head(10))

missing_df = resolved_df[resolved_df["Resolved_File"] == ""]
print("Missing resolved files:", len(missing_df))
if len(missing_df) > 0:
    display(missing_df)

def load_roi_table(fp):
    try:
        with open(fp, "r", encoding="utf-8", errors="ignore") as f:
            txt = f.read()
        tmp = pd.read_csv(StringIO(txt), sep="\t")
        if "Name" in tmp.columns:
            return tmp
    except Exception:
        pass

    for engine in [None, "xlrd", "openpyxl"]:
        try:
            tmp = pd.read_excel(fp, engine=engine)
            if "Name" in tmp.columns:
                return tmp
        except Exception:
            pass

    tmp = pd.read_csv(fp, sep=None, engine="python")
    if "Name" not in tmp.columns:
        raise ValueError("ROI table could not be parsed.")
    return tmp

roi_raw = load_roi_table(roi_path)
roi_df = roi_raw.copy()

if "ACTIVE" in roi_df.columns:
    roi_df["ACTIVE"] = roi_df["ACTIVE"].astype(str).str.upper()
    roi_df = roi_df[roi_df["ACTIVE"].isin(["TRUE", "T", "1"])].copy()

roi_df["ppm_left_base"] = roi_df[["w2_downfield", "w2_upfield"]].max(axis=1)
roi_df["ppm_right_base"] = roi_df[["w2_downfield", "w2_upfield"]].min(axis=1)
roi_df["ROI_Label"] = roi_df["Name"].astype(str)

ROI_RIGHT_EXPAND_PPM = {
    "ROI.2":  1.2,
    "ROI.10": 0.8,
    "ROI.11": 0.8,
    "ROI.13": 0.8,
}

roi_df["expand_right_ppm"] = roi_df["ROI_Label"].map(ROI_RIGHT_EXPAND_PPM).fillna(0.0)
roi_df["ppm_left"] = roi_df["ppm_left_base"]
roi_df["ppm_right"] = roi_df["ppm_right_base"] - roi_df["expand_right_ppm"]
roi_df["center"] = (roi_df["ppm_left"] + roi_df["ppm_right"]) / 2

EXCLUDE_ROIS = {"ROI.1", "ROI.15"}
roi_df = roi_df[~roi_df["ROI_Label"].isin(EXCLUDE_ROIS)].copy()

roi_df["roi_num"] = roi_df["ROI_Label"].str.extract(r"(\d+)").astype(int)
roi_df = roi_df.sort_values("roi_num").reset_index(drop=True)

if (roi_df["ROI_Label"] == "ROI.2").any():
    roi2_row = roi_df[roi_df["ROI_Label"] == "ROI.2"]
    roi_other = roi_df[roi_df["ROI_Label"] != "ROI.2"]
    roi_df = pd.concat([roi2_row, roi_other], axis=0).reset_index(drop=True)

roi_df.to_csv("roi_table_expanded.csv", index=False)

print("\nROI preview:")
display(roi_df[["ROI_Label","ppm_left_base","ppm_right_base","expand_right_ppm","ppm_left","ppm_right","center"]])

DEFAULT_COLOR = "#4D4D4D"
BLANK_COLOR = "#1565C0"

LINEWIDTH = 0.7
BLANK_LINEWIDTH = 1.2

ROI_FILL = "#F6B26B"
ROI_FILL_ALPHA = 0.20

ROI_PAD_FRAC = 0.04
MULTIWINDOW_NORMALIZATION_MODE = "per_window"

ROW_LABEL_FONTSIZE = 16
ROI_TITLE_FONTSIZE_BIG = 15

OVERLAY_AXIS_LABEL_FONTSIZE = 19
OVERLAY_TICK_FONTSIZE = 16
OVERLAY_TITLE_FONTSIZE = 23
OVERLAY_LEGEND_FONT_SIZE = 12

APPLY_NORMALIZATION = True
MEDIAN_KERNEL = 5
SG_WINDOW = 17
SG_POLY = 3
GAUSS_SIGMA = 0.9

DENOISE_SAMPLE_NOS = {4, 23}

FIXED_SHIFT_SAMPLE_NOS = {4}
FIXED_SHIFT_PPM = 2.8

PEAK_MIN_PROM_REL = 0.05
PEAK_MIN_HEIGHT_REL = 0.08
PEAK_MIN_DISTANCE_PTS = 3

PLOT_DIR = "roi_delta_plots"
os.makedirs(PLOT_DIR, exist_ok=True)

GRAPH_DIR = "roi_feature_heatmaps"
os.makedirs(GRAPH_DIR, exist_ok=True)

FEATURES_FOR_HEATMAP = [
    "main_peak_ppm",
    "main_peak_height",
    "auc",
    "n_peaks",
    "center_of_mass_ppm",
    "weighted_std_ppm",
    "skewness",
    "delta_ppm",
    "delta_height",
    "delta_auc",
    "delta_n_peaks",
    "ratio_height",
    "ratio_auc",
    "ratio_n_peaks",
]

def parse_sample_tokens(sample_name):
    toks = str(sample_name).split("_")
    hydrophilic = toks[0] if len(toks) > 0 else "NA"
    hydrophobic = toks[1] if len(toks) > 1 else "NA"
    crosslinker = toks[2] if len(toks) > 2 else "NA"
    return hydrophilic, hydrophobic, crosslinker

def normalize_signal(y):
    mx = np.max(np.abs(y))
    if mx > 0:
        return y / mx
    return y.copy()

def denoise_signal_no_baseline(y, median_kernel=5, sg_window=17, sg_poly=3, gaussian_sigma=0.9):
    y_work = y.astype(np.float64).copy()

    if median_kernel is not None and median_kernel >= 3:
        if median_kernel % 2 == 0:
            median_kernel += 1
        y_work = medfilt(y_work, kernel_size=median_kernel)

    if sg_window is not None and sg_window >= 5:
        if sg_window % 2 == 0:
            sg_window += 1
        if sg_window <= sg_poly:
            sg_window = sg_poly + 3 if (sg_poly + 3) % 2 == 1 else sg_poly + 4
        y_work = savgol_filter(y_work, window_length=sg_window, polyorder=sg_poly)

    if gaussian_sigma is not None and gaussian_sigma > 0:
        peak_mask = y_work > np.percentile(y_work, 85)
        y_smooth = gaussian_filter1d(y_work, sigma=gaussian_sigma)
        y_work = np.where(peak_mask, 0.7 * y_work + 0.3 * y_smooth, y_smooth)

    return y_work

def preprocess_spectrum(fp, normalize=True):
    dic, data, ppm = read_1d_ucsf_manual(fp)
    y = data.astype(np.float64)
    if normalize:
        y = normalize_signal(y)
    return ppm, y

def crop_roi(ppm, y, left_ppm, right_ppm, pad_frac=0.04):
    width = abs(left_ppm - right_ppm)
    pad = width * pad_frac
    lo = min(left_ppm, right_ppm) - pad
    hi = max(left_ppm, right_ppm) + pad
    mask = (ppm >= lo) & (ppm <= hi)
    return ppm[mask], y[mask], lo, hi

def safe_trapz(x, y):
    if len(x) < 2:
        return np.nan
    xs = np.asarray(x)
    ys = np.asarray(y)
    order = np.argsort(xs)
    return np.trapz(ys[order], xs[order])

def weighted_mean(x, w):
    wsum = np.sum(w)
    if wsum <= 0:
        return np.nan
    return np.sum(x * w) / wsum

def weighted_std(x, w, mu=None):
    wsum = np.sum(w)
    if wsum <= 0:
        return np.nan
    if mu is None:
        mu = weighted_mean(x, w)
    var = np.sum(w * (x - mu) ** 2) / wsum
    return np.sqrt(var)

def weighted_skewness(x, w, mu=None, sigma=None):
    wsum = np.sum(w)
    if wsum <= 0:
        return np.nan
    if mu is None:
        mu = weighted_mean(x, w)
    if sigma is None:
        sigma = weighted_std(x, w, mu=mu)
    if (sigma is None) or (np.isnan(sigma)) or (sigma == 0):
        return np.nan
    return np.sum(w * ((x - mu) / sigma) ** 3) / wsum

def quantify_roi(ppm_roi, y_roi):
    result = {
        "auc": np.nan,
        "max_intensity": np.nan,
        "center_of_mass_ppm": np.nan,
        "weighted_std_ppm": np.nan,
        "skewness": np.nan,
        "main_peak_ppm": np.nan,
        "main_peak_height": np.nan,
        "main_peak_prominence": np.nan,
        "main_peak_width_ppm": np.nan,
        "n_peaks": 0,
    }

    if len(ppm_roi) < 5:
        return result

    x = np.asarray(ppm_roi, dtype=float)
    y = np.asarray(y_roi, dtype=float)
    y_pos = y.copy()
    y_pos[y_pos < 0] = 0.0

    if np.allclose(y_pos, 0):
        return result

    result["auc"] = safe_trapz(x, y_pos)
    result["max_intensity"] = float(np.max(y_pos))

    mu = weighted_mean(x, y_pos)
    sd = weighted_std(x, y_pos, mu=mu)
    sk = weighted_skewness(x, y_pos, mu=mu, sigma=sd)

    result["center_of_mass_ppm"] = mu
    result["weighted_std_ppm"] = sd
    result["skewness"] = sk

    ymax = np.max(y_pos)
    prom_thr = PEAK_MIN_PROM_REL * ymax
    h_thr = PEAK_MIN_HEIGHT_REL * ymax

    peaks, props = find_peaks(
        y_pos,
        prominence=prom_thr,
        height=h_thr,
        distance=PEAK_MIN_DISTANCE_PTS
    )

    result["n_peaks"] = int(len(peaks))

    if len(peaks) > 0:
        prominences = props.get("prominences", np.full(len(peaks), np.nan))
        heights = props.get("peak_heights", y_pos[peaks])

        main_idx = int(np.nanargmax(prominences))
        peak_i = peaks[main_idx]

        result["main_peak_ppm"] = float(x[peak_i])
        result["main_peak_height"] = float(heights[main_idx])
        result["main_peak_prominence"] = float(prominences[main_idx])

        try:
            widths, _, left_ips, right_ips = peak_widths(y_pos, [peak_i], rel_height=0.5)
            grid = np.arange(len(x))
            left_ppm = np.interp(left_ips[0], grid, x)
            right_ppm = np.interp(right_ips[0], grid, x)
            result["main_peak_width_ppm"] = float(abs(right_ppm - left_ppm))
        except Exception:
            result["main_peak_width_ppm"] = np.nan

    return result

def safe_ratio(x, y):
    if y is None or not np.isfinite(y) or y == 0:
        return np.nan
    return x / y

def set_bottom_ppm_labels_outside(ax, left_ppm, right_ppm, center_ppm=None):
    ax.set_xticks([])
    ax.tick_params(axis="x", length=0, labelbottom=False)

def add_roi_title_selected(ax, roi_label):
    ax.set_title(
        roi_label,
        fontsize=ROI_TITLE_FONTSIZE_BIG,
        fontweight="bold",
        pad=6,
        color="black"
    )

def draw_overlay_roi_spans(ax, roi_table):
    for _, r in roi_table.iterrows():
        ax.axvspan(
            float(r["ppm_left"]),
            float(r["ppm_right"]),
            color=ROI_FILL,
            alpha=ROI_FILL_ALPHA,
            zorder=0
        )

def build_multiwindow_label(no, item):
    if no == 4:
        return f"{no}. {item['sample_name']}"
    if no == 23:
        return f"{no}. {item['sample_name']}"
    tag = " [den]" if "denoised" in item["processing"] else ""
    if no in FIXED_SHIFT_SAMPLE_NOS:
        tag += f" [shift+{FIXED_SHIFT_PPM:.1f}]"
    return f"{no}. {item['sample_name']}{tag}"

def sanitize_feature_name(s):
    return str(s).replace("/", "_").replace(" ", "_")

def plot_feature_heatmap(df_long, feature, out_png):
    pivot = df_long.pivot(index="No", columns="ROI_Label", values=feature)
    if pivot.shape[0] == 0 or pivot.shape[1] == 0:
        return None

    cols_sorted = sorted(pivot.columns.tolist(), key=roi_sort_key)
    pivot = pivot[cols_sorted].sort_index()

    arr = pivot.to_numpy(dtype=float)
    arr_masked = np.ma.masked_invalid(arr)

    fig_w = max(10, 0.72 * pivot.shape[1] + 3)
    fig_h = max(8, 0.28 * pivot.shape[0] + 2.5)

    fig, ax = plt.subplots(figsize=(fig_w, fig_h))
    im = ax.imshow(arr_masked, aspect="auto", interpolation="nearest")

    ax.set_xticks(np.arange(pivot.shape[1]))
    ax.set_xticklabels(pivot.columns, rotation=45, ha="right", fontsize=9)
    ax.set_yticks(np.arange(pivot.shape[0]))
    ax.set_yticklabels([str(x) for x in pivot.index], fontsize=8)

    ax.set_xlabel("ROI")
    ax.set_ylabel("Sample No.")
    ax.set_title(f"Heatmap of {feature}")

    cbar = plt.colorbar(im, ax=ax)
    cbar.set_label(feature)

    plt.tight_layout()
    plt.savefig(out_png, dpi=300, bbox_inches="tight")
    plt.close(fig)
    return out_png

blank_ppm, blank_y = preprocess_spectrum(blank_path, normalize=APPLY_NORMALIZATION)

spec_cache = {}
spec_cache[0] = {
    "No": 0,
    "sample_name": "Blank",
    "crosslinker": "Blank",
    "hydrophilic": "Blank",
    "hydrophobic": "Blank",
    "ppm": blank_ppm,
    "y": blank_y,
    "processing": "raw",
    "file": blank_path
}

export_rows = []

for _, r in resolved_df.iterrows():
    if r["Resolved_File"] == "":
        continue

    no = int(r["No"])
    sample_name = r["Sample_Name"]
    fp = r["Resolved_File"]
    hydrophilic, hydrophobic, crosslinker = parse_sample_tokens(sample_name)

    ppm, y = preprocess_spectrum(fp, normalize=APPLY_NORMALIZATION)

    if no in DENOISE_SAMPLE_NOS:
        y_use = denoise_signal_no_baseline(
            y,
            median_kernel=MEDIAN_KERNEL,
            sg_window=SG_WINDOW,
            sg_poly=SG_POLY,
            gaussian_sigma=GAUSS_SIGMA
        )
        processing = "denoised"
    else:
        y_use = y.copy()
        processing = "raw"

    if no in FIXED_SHIFT_SAMPLE_NOS:
        ppm_use = ppm + FIXED_SHIFT_PPM
        processing = f"{processing}_ppmshift(+{FIXED_SHIFT_PPM:.1f})"
    else:
        ppm_use = ppm.copy()

    spec_cache[no] = {
        "No": no,
        "sample_name": sample_name,
        "crosslinker": crosslinker,
        "hydrophilic": hydrophilic,
        "hydrophobic": hydrophobic,
        "ppm": ppm_use,
        "y": y_use,
        "processing": processing,
        "file": fp
    }

    export_rows.append(pd.DataFrame({
        "No": no,
        "Sample_Name": sample_name,
        "ppm_original": ppm,
        "ppm_used": ppm_use,
        "signal_used": y_use,
        "processing": processing
    }))

if len(export_rows) > 0:
    pd.concat(export_rows, axis=0, ignore_index=True).to_csv(
        "all_samples_selected_processing_long.csv", index=False
    )

print("\nCached samples:", len(spec_cache))
print("Denoised samples:", sorted(DENOISE_SAMPLE_NOS))
print("Shifted samples :", sorted(FIXED_SHIFT_SAMPLE_NOS), f"(+{FIXED_SHIFT_PPM:.1f} ppm)")

def plot_overlay_for_subset(sample_nos, title, out_stem, alpha=0.65, show_legend=True, add_roi_spans=False):
    plt.figure(figsize=(13.5, 7.0))
    ax = plt.gca()

    if add_roi_spans:
        draw_overlay_roi_spans(ax, roi_df)

    handles = []
    labels = []

    for no in sample_nos:
        if no not in spec_cache:
            continue
        item = spec_cache[no]

        h, = ax.plot(
            item["ppm"], item["y"],
            color=DEFAULT_COLOR,
            lw=LINEWIDTH,
            alpha=alpha,
            label=f"{no}. {item['sample_name']}"
        )
        handles.append(h)
        labels.append(f"{no}. {item['sample_name']}")

    h_blank, = ax.plot(
        blank_ppm,
        blank_y,
        color=BLANK_COLOR,
        lw=BLANK_LINEWIDTH,
        alpha=1.0,
        label="Blank"
    )
    handles.append(h_blank)
    labels.append("Blank")

    ax.invert_xaxis()

    # ===== ここだけ変更：Y軸目盛を 0, 0.5, 1.0 に固定 =====
    ax.set_xlabel("¹H chemical shift (ppm)", fontsize=OVERLAY_AXIS_LABEL_FONTSIZE)
    ax.set_ylabel("Intensity", fontsize=OVERLAY_AXIS_LABEL_FONTSIZE)

    ax.set_ylim(0, 1.0)
    ax.set_yticks([0, 0.5, 1.0])

    ax.tick_params(axis="both", labelsize=OVERLAY_TICK_FONTSIZE)
    # ======================================================

    ax.set_title(title, fontsize=OVERLAY_TITLE_FONTSIZE, pad=12)
    ax.grid(False)

    if show_legend:
        ax.legend(
            handles,
            labels,
            fontsize=OVERLAY_LEGEND_FONT_SIZE,
            frameon=False,
            ncol=1,
            loc="upper left",
            bbox_to_anchor=(0.005, 0.995),
            borderaxespad=0.2
        )

    plt.tight_layout()
    plt.savefig(f"{out_stem}.png", dpi=300, bbox_inches="tight")
    plt.savefig(f"{out_stem}.pdf", bbox_inches="tight")
    plt.show()

all_sample_nos = [int(x) for x in resolved_df["No"].tolist() if int(x) in spec_cache]

overlay_groups = {
    "1_11":  [x for x in all_sample_nos if 1  <= x <= 11],
    "12_22": [x for x in all_sample_nos if 12 <= x <= 22],
    "23_33": [x for x in all_sample_nos if 23 <= x <= 33],
    "34_43": [x for x in all_sample_nos if 34 <= x <= 43],
}

for key, subset in overlay_groups.items():
    plot_overlay_for_subset(
        subset,
        f"1D zgpg30 overlay: {key.replace('_','-')} + Blank",
        f"overlay_{key}_blank",
        alpha=0.72,
        show_legend=True,
        add_roi_spans=True
    )

plot_overlay_for_subset(
    all_sample_nos,
    "Direct overlay of 1D zgpg30 + Blank",
    "overlay_all_samples_blank",
    alpha=0.35,
    show_legend=False,
    add_roi_spans=True
)

sample_list = [{"No": 0, "Label": "Blank", "ppm": blank_ppm, "y": blank_y}]
for no in all_sample_nos:
    item = spec_cache[no]
    sample_list.append({
        "No": no,
        "Label": build_multiwindow_label(no, item),
        "ppm": item["ppm"],
        "y": item["y"]
    })

n_rows = len(sample_list)
n_cols = len(roi_df)

fig_w = max(12, 1.55 * n_cols)
fig_h = max(9, 0.65 * n_rows + 1.8)

fig, axes = plt.subplots(
    n_rows,
    n_cols,
    figsize=(fig_w, fig_h),
    squeeze=False,
    sharex=False,
    sharey=False,
    facecolor="white"
)

for row_idx, item in enumerate(sample_list):
    ppm = item["ppm"]
    y_global = item["y"]
    label = item["Label"]

    for col_idx, (_, rr) in enumerate(roi_df.iterrows()):
        ax = axes[row_idx, col_idx]

        left_ppm = float(rr["ppm_left"])
        right_ppm = float(rr["ppm_right"])
        center_ppm = float(rr["center"])

        ppm_roi, y_roi, lo, hi = crop_roi(ppm, y_global, left_ppm, right_ppm, ROI_PAD_FRAC)

        if len(ppm_roi) == 0:
            ax.axis("off")
            continue

        y_plot = normalize_signal(y_roi) if MULTIWINDOW_NORMALIZATION_MODE == "per_window" else y_roi.copy()

        ax.plot(ppm_roi, y_plot, color="black", lw=0.7)

        ax.set_facecolor("white")
        for spine in ax.spines.values():
            spine.set_color("black")
            spine.set_linewidth(0.6)

        ax.set_xlim(max(lo, hi), min(lo, hi))
        ax.set_ylim(-0.05, 1.20)
        ax.set_yticks([])
        ax.tick_params(axis="y", length=0)

        ax.set_xticks([])
        ax.tick_params(axis="x", length=0, labelbottom=False)

        if row_idx == 0:
            add_roi_title_selected(ax, rr["ROI_Label"])

        if col_idx == 0:
            ax.set_ylabel(
                label,
                fontsize=ROW_LABEL_FONTSIZE,
                rotation=0,
                ha="right",
                va="center",
                labelpad=28,
                color="black",
                fontweight="bold"
            )

plt.subplots_adjust(
    left=0.23,
    right=0.995,
    top=0.91,
    bottom=0.08,
    wspace=0.035,
    hspace=0.04
)

fig.suptitle(
    "ROI-wise multi-window map",
    fontsize=17,
    y=0.97
)
plt.savefig("roi_multiwindow_map_selective_denoise.png", dpi=300, bbox_inches="tight", facecolor="white")
plt.savefig("roi_multiwindow_map_selective_denoise.pdf", bbox_inches="tight", facecolor="white")
plt.show()

records = []

for no, item in spec_cache.items():
    ppm = item["ppm"]
    y = item["y"]

    for _, rr in roi_df.iterrows():
        roi_label = rr["ROI_Label"]
        left_ppm = float(rr["ppm_left"])
        right_ppm = float(rr["ppm_right"])
        center_ppm = float(rr["center"])

        ppm_roi, y_roi, _, _ = crop_roi(ppm, y, left_ppm, right_ppm, ROI_PAD_FRAC)
        feats = quantify_roi(ppm_roi, y_roi)

        records.append({
            "No": item["No"],
            "Sample_Name": item["sample_name"],
            "Hydrophilic": item["hydrophilic"],
            "Hydrophobic": item["hydrophobic"],
            "Crosslinker": item["crosslinker"],
            "Processing": item["processing"],
            "ROI_Label": roi_label,
            "ROI_Left_ppm": left_ppm,
            "ROI_Right_ppm": right_ppm,
            "ROI_Center_ppm": center_ppm,
            **feats
        })

roi_feat_df = pd.DataFrame(records).sort_values(["ROI_Label", "No"]).reset_index(drop=True)
roi_feat_df.to_csv("roi_peak_features_long.csv", index=False)

blank_df = roi_feat_df[roi_feat_df["No"] == 0].copy()
blank_dict = {r["ROI_Label"]: r for _, r in blank_df.iterrows()}

records2 = []
for _, r in roi_feat_df.iterrows():
    roi = r["ROI_Label"]
    b = blank_dict[roi]

    rec = r.to_dict()
    rec["delta_height"] = r["main_peak_height"] - b["main_peak_height"]
    rec["ratio_height"] = safe_ratio(r["main_peak_height"], b["main_peak_height"])
    rec["delta_auc"] = r["auc"] - b["auc"]
    rec["ratio_auc"] = safe_ratio(r["auc"], b["auc"])
    rec["delta_n_peaks"] = r["n_peaks"] - b["n_peaks"]
    rec["ratio_n_peaks"] = safe_ratio(r["n_peaks"], b["n_peaks"])
    rec["delta_ppm"] = r["main_peak_ppm"] - b["main_peak_ppm"]
    records2.append(rec)

roi_delta_df = pd.DataFrame(records2).sort_values(["ROI_Label", "No"]).reset_index(drop=True)
roi_delta_df.to_csv("roi_peak_features_with_blank_relative.csv", index=False)

feature_cols = [
    "main_peak_height", "auc", "n_peaks", "main_peak_ppm",
    "delta_height", "ratio_height",
    "delta_auc", "ratio_auc",
    "delta_n_peaks", "ratio_n_peaks",
    "delta_ppm"
]

base_meta = roi_delta_df[["No", "Sample_Name", "Hydrophilic", "Hydrophobic", "Crosslinker", "Processing"]].drop_duplicates()
wide_df = base_meta.copy()

for feat in feature_cols:
    tmp = roi_delta_df.pivot(index="No", columns="ROI_Label", values=feat)
    tmp.columns = [f"{c}__{feat}" for c in tmp.columns]
    tmp = tmp.reset_index()
    wide_df = wide_df.merge(tmp, on="No", how="left")

wide_df = wide_df.sort_values("No").reset_index(drop=True)
wide_df.to_csv("roi_peak_features_with_blank_relative_wide.csv", index=False)

sample_only_df = roi_delta_df[roi_delta_df["No"] != 0].copy()

roi_intensity_wide = sample_only_df.pivot_table(
    index=["No", "Sample_Name"],
    columns="ROI_Label",
    values="main_peak_height",
    aggfunc="first"
).reset_index()

roi_cols_sorted = sorted(
    [c for c in roi_intensity_wide.columns if str(c).startswith("ROI.")],
    key=roi_sort_key
)

roi_intensity_wide = roi_intensity_wide[["No", "Sample_Name"] + roi_cols_sorted].copy()
roi_intensity_wide = roi_intensity_wide.rename(columns={"Sample_Name": "Copolymer_Name"})
roi_intensity_wide["n_nonmissing_roi_values"] = roi_intensity_wide[roi_cols_sorted].notna().sum(axis=1)
roi_intensity_wide["mean_abs_roi_intensity"] = roi_intensity_wide[roi_cols_sorted].abs().mean(axis=1)

roi_intensity_compact = roi_intensity_wide[["Copolymer_Name"] + roi_cols_sorted].copy()

OUT_INT_CSV  = "ROIintensity_zgpg30_260423.csv"
OUT_INT_XLSX = "ROIintensity_zgpg30_260423.xlsx"

roi_intensity_compact.to_csv(OUT_INT_CSV, index=False, encoding="utf-8-sig")

with pd.ExcelWriter(OUT_INT_XLSX, engine="xlsxwriter") as writer:
    roi_intensity_compact.to_excel(writer, index=False, sheet_name="ROIintensity")
    roi_intensity_wide.to_excel(writer, index=False, sheet_name="ROIintensity_with_meta")

roi_auc_wide = sample_only_df.pivot_table(
    index=["No", "Sample_Name"],
    columns="ROI_Label",
    values="auc",
    aggfunc="first"
).reset_index()

roi_auc_cols_sorted = sorted(
    [c for c in roi_auc_wide.columns if str(c).startswith("ROI.")],
    key=roi_sort_key
)

roi_auc_wide = roi_auc_wide[["No", "Sample_Name"] + roi_auc_cols_sorted].copy()
roi_auc_wide = roi_auc_wide.rename(columns={"Sample_Name": "Copolymer_Name"})
roi_auc_wide["n_nonmissing_roi_values"] = roi_auc_wide[roi_auc_cols_sorted].notna().sum(axis=1)
roi_auc_wide["mean_abs_roi_auc"] = roi_auc_wide[roi_auc_cols_sorted].abs().mean(axis=1)

roi_auc_compact = roi_auc_wide[["Copolymer_Name"] + roi_auc_cols_sorted].copy()

OUT_AUC_CSV  = "ROIauc_zgpg30_260423.csv"
OUT_AUC_XLSX = "ROIauc_zgpg30_260423.xlsx"

roi_auc_compact.to_csv(OUT_AUC_CSV, index=False, encoding="utf-8-sig")

with pd.ExcelWriter(OUT_AUC_XLSX, engine="xlsxwriter") as writer:
    roi_auc_compact.to_excel(writer, index=False, sheet_name="ROIauc")
    roi_auc_wide.to_excel(writer, index=False, sheet_name="ROIauc_with_meta")

print("\nSaved additional files:")
print(" -", OUT_INT_CSV)
print(" -", OUT_INT_XLSX)
print(" -", OUT_AUC_CSV)
print(" -", OUT_AUC_XLSX)

generated_graphs = []

for feat in FEATURES_FOR_HEATMAP:
    out_png = os.path.join(GRAPH_DIR, f"heatmap_{sanitize_feature_name(feat)}.png")
    fp = plot_feature_heatmap(roi_delta_df, feat, out_png)
    if fp is not None:
        generated_graphs.append((fp, f"Heatmap: {feat}"))

print("\nGenerated quantified-feature heatmaps:")
for fp, ttl in generated_graphs:
    print(" -", fp, "|", ttl)

def add_title(slide, title, subtitle=""):
    tx = slide.shapes.add_textbox(Inches(0.4), Inches(0.2), Inches(12.0), Inches(0.5))
    tf = tx.text_frame
    p = tf.paragraphs[0]
    r = p.add_run()
    r.text = title
    r.font.size = Pt(22)
    r.font.bold = True

    if subtitle:
        tx2 = slide.shapes.add_textbox(Inches(0.4), Inches(0.65), Inches(12.0), Inches(0.3))
        tf2 = tx2.text_frame
        p2 = tf2.paragraphs[0]
        r2 = p2.add_run()
        r2.text = subtitle
        r2.font.size = Pt(10)

def add_picture_fit(slide, img_path, x, y, w, h):
    slide.shapes.add_picture(img_path, Inches(x), Inches(y), Inches(w), Inches(h))

prs = Presentation()
prs.slide_width = Inches(13.333)
prs.slide_height = Inches(7.5)

slide = prs.slides.add_slide(prs.slide_layouts[6])
add_title(
    slide,
    "ROI-expanded quantification report",
    "Overlay plots + multi-window map + quantified ROI heatmaps"
)

overlay_files = [
    ("overlay_1_11_blank.png", "1-11 + Blank"),
    ("overlay_12_22_blank.png", "12-22 + Blank"),
    ("overlay_23_33_blank.png", "23-33 + Blank"),
    ("overlay_34_43_blank.png", "34-43 + Blank"),
    ("overlay_all_samples_blank.png", "All samples + Blank"),
    ("roi_multiwindow_map_selective_denoise.png", "ROI multi-window map"),
]

for fp, ttl in overlay_files:
    if os.path.exists(fp):
        slide = prs.slides.add_slide(prs.slide_layouts[6])
        add_title(slide, ttl)
        add_picture_fit(slide, fp, 0.4, 1.0, 12.4, 6.0)

for fp, ttl in generated_graphs:
    if os.path.exists(fp):
        slide = prs.slides.add_slide(prs.slide_layouts[6])
        add_title(slide, ttl)
        add_picture_fit(slide, fp, 0.4, 1.0, 12.4, 6.0)

ppt_path = "roi_quantification_report_with_heatmaps.pptx"
prs.save(ppt_path)

output_files = [
    "resolved_zgpg30_1d.csv",
    "roi_table_expanded.csv",
    "all_samples_selected_processing_long.csv",
    "overlay_1_11_blank.png", "overlay_1_11_blank.pdf",
    "overlay_12_22_blank.png", "overlay_12_22_blank.pdf",
    "overlay_23_33_blank.png", "overlay_23_33_blank.pdf",
    "overlay_34_43_blank.png", "overlay_34_43_blank.pdf",
    "overlay_all_samples_blank.png", "overlay_all_samples_blank.pdf",
    "roi_multiwindow_map_selective_denoise.png",
    "roi_multiwindow_map_selective_denoise.pdf",
    "roi_peak_features_long.csv",
    "roi_peak_features_with_blank_relative.csv",
    "roi_peak_features_with_blank_relative_wide.csv",
    "ROIintensity_zgpg30_260423.csv",
    "ROIintensity_zgpg30_260423.xlsx",
    "ROIauc_zgpg30_260423.csv",
    "ROIauc_zgpg30_260423.xlsx",
    ppt_path,
]

for fp, _ in generated_graphs:
    output_files.append(fp)

zip_name = "zgpg30_roi_outputs_with_intensity_auc_260423_modified.zip"
with zipfile.ZipFile(zip_name, "w", zipfile.ZIP_DEFLATED) as zf:
    for fp in output_files:
        if os.path.exists(fp):
            zf.write(fp)

print("\nSaved files:")
for fp in output_files:
    if os.path.exists(fp):
        print(" -", fp)

print(" -", zip_name)

files.download("ROIintensity_zgpg30_260423.csv")
files.download("ROIintensity_zgpg30_260423.xlsx")
files.download("ROIauc_zgpg30_260423.csv")
files.download("ROIauc_zgpg30_260423.xlsx")
files.download(ppt_path)
files.download(zip_name)
```
