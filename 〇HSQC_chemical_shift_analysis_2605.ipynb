```python
# ============================================================
# Google Colab complete code
# HSQC visualization pipeline
#
# Modified points:
#   1) Overlay groups:
#        Blank + No.1–11
#        Blank + No.12–23
#        Blank + No.24–33
#        Blank + No.34–43
#   2) Overlay keeps all ROIs.
#   3) Multi-window ROI numbers are re-indexed from ROI.1.
# ============================================================

!pip -q install nmrglue pandas matplotlib numpy openpyxl xlrd pillow

import os
import re
import glob
import zipfile
import shutil
from pathlib import Path
from io import StringIO

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import nmrglue as ng

from google.colab import files
uploaded = files.upload()

# ------------------------------------------------------------
# 0. File detection
# ------------------------------------------------------------
def find_first(patterns):
    hits = []
    for p in patterns:
        hits.extend(glob.glob(p))
    hits = sorted(set(hits))
    return hits[0] if hits else None

zip_path = find_first(["ucsf(2).zip", "ucsf*.zip", "*.zip"])
csv_path = find_first(["sample_1_43_ucsf_mapping*.csv", "*mapping*.csv"])
blank_path = find_first(["Blank602_HSQC*.ucsf", "Blank*HSQC*.ucsf", "Blank*.ucsf"])
roi_path = find_first([
    "roiTable_hsqc*.txt", "roiTable_hsqc*.csv",
    "roiTable_hsqc*.xls", "roiTable_hsqc*.xlsx",
    "*hsqc*roi*.txt", "*hsqc*roi*.csv",
    "*roi*.txt", "*roi*.csv", "*roi*.xls", "*roi*.xlsx"
])

if zip_path is None:
    raise FileNotFoundError("ZIP file not found.")
if csv_path is None:
    raise FileNotFoundError("Mapping CSV not found.")
if blank_path is None:
    raise FileNotFoundError("HSQC blank UCSF file not found.")
if roi_path is None:
    raise FileNotFoundError("HSQC ROI table file not found.")

print("ZIP   :", zip_path)
print("CSV   :", csv_path)
print("BLANK :", blank_path)
print("ROI   :", roi_path)

# ------------------------------------------------------------
# 1. Extract ZIP
# ------------------------------------------------------------
extract_dir = "extracted_hsqc_ucsf"
if os.path.exists(extract_dir):
    shutil.rmtree(extract_dir)
os.makedirs(extract_dir, exist_ok=True)

with zipfile.ZipFile(zip_path, "r") as zf:
    zf.extractall(extract_dir)

all_ucsf_files = sorted(glob.glob(os.path.join(extract_dir, "**", "*.ucsf"), recursive=True))
all_ucsf_files += glob.glob("*.ucsf")
all_ucsf_files = sorted(set(all_ucsf_files))

print("Total UCSF candidates:", len(all_ucsf_files))
for x in all_ucsf_files[:20]:
    print(" -", os.path.basename(x))

# ------------------------------------------------------------
# 2. Read mapping CSV
# ------------------------------------------------------------
df = pd.read_csv(csv_path)

required_cols = ["No", "Sample_Name", "Component_Number"]
for c in required_cols:
    if c not in df.columns:
        raise ValueError(f"Required column not found: {c}")

df = df.sort_values("No").reset_index(drop=True)
df["No"] = pd.to_numeric(df["No"], errors="coerce").astype("Int64")
df["Sample_Name"] = df["Sample_Name"].astype(str).str.strip()
df["Component_Number"] = (
    df["Component_Number"].astype(str).str.strip().str.replace("-", "_", regex=False)
)

optional_number_cols = []
for c in ["HSQC_ucsf_No", "hsqc_ucsf_No", "hsqc_ucsf_no", "zgpg30_ucsf_No"]:
    if c in df.columns:
        optional_number_cols.append(c)

print("Optional numeric resolver columns:", optional_number_cols)
display(df.head())

# ------------------------------------------------------------
# 3. Read 2D UCSF
# ------------------------------------------------------------
def read_2d_ucsf(fp):
    dic, data = ng.sparky.read(fp)
    if data.ndim != 2:
        raise ValueError(f"Not a 2D UCSF file: {fp}")

    uc0 = ng.sparky.make_uc(dic, data, dim=0)
    uc1 = ng.sparky.make_uc(dic, data, dim=1)

    ppm0 = uc0.ppm_scale()
    ppm1 = uc1.ppm_scale()

    return dic, np.asarray(data, dtype=np.float64), ppm0, ppm1

def try_read_2d(fp):
    try:
        return read_2d_ucsf(fp)
    except Exception:
        return None

readable_2d_files = [fp for fp in all_ucsf_files if try_read_2d(fp) is not None]

print("Readable 2D UCSF files:", len(readable_2d_files))
for x in readable_2d_files[:20]:
    print(" -", os.path.basename(x))

# ------------------------------------------------------------
# 4. Resolver
# ------------------------------------------------------------
MANUAL_HSQC_FILE_MAP = {
    # Example:
    # 7: "1513_HSQC.ucsf"
}

def stem_no_ext(path):
    return Path(path).stem

def normalize_key(s):
    s = str(s).strip()
    s = s.replace("-", "_")
    s = re.sub(r"\s+", "_", s)
    return s

def numeric_string(val):
    if pd.isna(val):
        return None
    try:
        return str(int(float(val)))
    except Exception:
        s = str(val).strip()
        return s if s else None

def token_numeric_match(stem, token):
    if token is None:
        return False
    return re.search(rf"(^|[_\-\s]){re.escape(token)}($|[_\-\s])", stem) is not None

def file_priority_score(fp):
    low = fp.lower()
    score = 100
    if "hsqc" in low:
        score -= 50
    if "blank" in low:
        score += 500
    if "zgpg30" in low:
        score += 1000
    return score

file_meta = []
for fp in readable_2d_files:
    stem = stem_no_ext(fp)
    norm = normalize_key(stem)
    nums = re.findall(r"\d+", stem)
    file_meta.append({
        "path": fp,
        "basename": os.path.basename(fp),
        "stem": stem,
        "norm": norm,
        "nums": nums
    })

def choose_best(cands):
    if not cands:
        return None
    return sorted(set(cands), key=file_priority_score)[0]

def resolve_manual_file(no):
    if no not in MANUAL_HSQC_FILE_MAP:
        return None, ""
    target = MANUAL_HSQC_FILE_MAP[no]
    for fm in file_meta:
        if fm["basename"] == target:
            return fm["path"], "manual_override"
    return None, ""

def resolve_hsqc_file(row):
    no = int(row["No"])
    sname = normalize_key(row["Sample_Name"])
    comp = normalize_key(row["Component_Number"])

    numeric_candidates = []
    for c in optional_number_cols:
        val = numeric_string(row[c])
        if val is not None:
            numeric_candidates.append((c, val))
    numeric_candidates.append(("No", str(no)))

    fp, reason = resolve_manual_file(no)
    if fp is not None:
        return fp, reason

    for src, num in numeric_candidates:
        cands = [fm["path"] for fm in file_meta if fm["stem"] == num or fm["norm"] == num]
        fp = choose_best(cands)
        if fp is not None:
            return fp, f"{src}_exact"

    for src, num in numeric_candidates:
        cands = [fm["path"] for fm in file_meta if token_numeric_match(fm["stem"], num)]
        fp = choose_best(cands)
        if fp is not None:
            return fp, f"{src}_token"

    cands = [fm["path"] for fm in file_meta if fm["norm"] == sname]
    fp = choose_best(cands)
    if fp is not None:
        return fp, "sample_exact"

    cands = [fm["path"] for fm in file_meta if sname in fm["norm"]]
    fp = choose_best(cands)
    if fp is not None:
        return fp, "sample_contains"

    cands = [fm["path"] for fm in file_meta if fm["norm"] == comp]
    fp = choose_best(cands)
    if fp is not None:
        return fp, "component_exact"

    cands = [fm["path"] for fm in file_meta if fm["norm"].startswith(comp)]
    fp = choose_best(cands)
    if fp is not None:
        return fp, "component_prefix"

    cands = [fm["path"] for fm in file_meta if comp in fm["norm"]]
    fp = choose_best(cands)
    if fp is not None:
        return fp, "component_contains"

    return None, "unresolved"

resolved_records = []
for _, row in df.iterrows():
    fp, reason = resolve_hsqc_file(row)
    resolved_records.append({
        "No": int(row["No"]),
        "Sample_Name": row["Sample_Name"],
        "Component_Number": row["Component_Number"],
        "Resolved_File": fp if fp is not None else "",
        "Resolved_Basename": os.path.basename(fp) if fp is not None else "",
        "Resolver_Reason": reason
    })

resolved_df = pd.DataFrame(resolved_records)
display(resolved_df)

resolved_count = (resolved_df["Resolved_File"] != "").sum()
print(f"Resolved count: {resolved_count} / {len(resolved_df)}")

missing_df = resolved_df[resolved_df["Resolved_File"] == ""]
print("Missing resolved files:", len(missing_df))
if len(missing_df) > 0:
    display(missing_df)

resolved_df.to_csv("resolved_hsqc_2d_files_code3_style.csv", index=False)

# ------------------------------------------------------------
# 5. Load ROI table
# ------------------------------------------------------------
def load_roi_table(fp):
    try:
        with open(fp, "r", encoding="utf-8", errors="ignore") as f:
            txt = f.read()
        tmp = pd.read_csv(StringIO(txt), sep="\t")
        if "Name" in tmp.columns:
            return tmp
    except Exception:
        pass

    for engine in [None, "xlrd", "openpyxl"]:
        try:
            tmp = pd.read_excel(fp, engine=engine)
            if "Name" in tmp.columns:
                return tmp
        except Exception:
            pass

    tmp = pd.read_csv(fp, sep=None, engine="python")
    if "Name" not in tmp.columns:
        raise ValueError("ROI table could not be parsed.")
    return tmp

roi_raw = load_roi_table(roi_path)
roi_df = roi_raw.copy()

if "ACTIVE" in roi_df.columns:
    roi_df["ACTIVE"] = roi_df["ACTIVE"].astype(str).str.upper().str.strip()
    roi_df = roi_df[roi_df["ACTIVE"].isin(["TRUE", "T", "1", "YES"])].copy()

required_roi_cols = ["Name", "w2_downfield", "w2_upfield", "w1_downfield", "w1_upfield"]
for c in required_roi_cols:
    if c not in roi_df.columns:
        raise ValueError(f"ROI table must contain column: {c}")

for c in ["w2_downfield", "w2_upfield", "w1_downfield", "w1_upfield"]:
    roi_df[c] = pd.to_numeric(roi_df[c], errors="coerce")

roi_df = roi_df.dropna(subset=required_roi_cols).copy()

roi_df["Name"] = roi_df["Name"].astype(str).str.strip()
roi_df["w2_left"]   = roi_df[["w2_downfield", "w2_upfield"]].max(axis=1)
roi_df["w2_right"]  = roi_df[["w2_downfield", "w2_upfield"]].min(axis=1)
roi_df["w1_top"]    = roi_df[["w1_downfield", "w1_upfield"]].max(axis=1)
roi_df["w1_bottom"] = roi_df[["w1_downfield", "w1_upfield"]].min(axis=1)
roi_df["ROI_Label"] = roi_df["Name"]
roi_df = roi_df.reset_index(drop=True)

# ------------------------------------------------------------
# 5.1 ROI handling
# ------------------------------------------------------------
EXCLUDE_ROIS_MULTIWINDOW = {
    "ROI.2", "ROI.7", "ROI.13", "ROI.14", "ROI.15",
    "ROI.19", "ROI.28", "ROI.34", "ROI.35", "ROI.36"
}

# Overlay: keep all original ROIs
roi_df_overlay = roi_df.copy().reset_index(drop=True)

# Multi-window: remove selected ROIs
roi_df_multi = roi_df[
    ~roi_df["ROI_Label"].isin(EXCLUDE_ROIS_MULTIWINDOW)
].copy().reset_index(drop=True)

# Re-number ROI labels from ROI.1 for Multi-window figure
roi_df_multi["Original_ROI_Label"] = roi_df_multi["ROI_Label"]
roi_df_multi["ROI_Label"] = [f"ROI.{i+1}" for i in range(len(roi_df_multi))]

print("Overlay ROI count:", len(roi_df_overlay))
print("Multi-window ROI count:", len(roi_df_multi))
print("Removed ROIs:", sorted(EXCLUDE_ROIS_MULTIWINDOW))
print("Multi-window ROI relabeling:")
display(roi_df_multi[[
    "Original_ROI_Label", "ROI_Label",
    "w2_left", "w2_right", "w1_top", "w1_bottom"
]])

# ------------------------------------------------------------
# 6. Settings
# ------------------------------------------------------------
BLANK_COLOR = "black"

CROSSLINKER_COLORS = {
    "TECL": "#A5D6A7",
    "TRCL": "#43A047",
    "DICL": "#1B5E20",
}
DEFAULT_COLOR = "#4D4D4D"
SAMPLE_CONTOUR_COLOR = "black"
ROI_BOX_COLOR = "#0B3D91"

N_LEVELS = 8
LEVEL_MULTIPLIER = 1.40
BASE_NOISE_Q = 98.5

ROI_PAD_FRAC_X = 0.05
ROI_PAD_FRAC_Y = 0.05

ROW_HEIGHT = 0.62
COL_WIDTH = 1.55

ROW_LABEL_FONTSIZE = 13
ROI_TITLE_FONTSIZE = 13
PPM_FONTSIZE = 9

OVERLAY_TITLE_FONTSIZE = 15
AXIS_LABEL_FONTSIZE = 12
SUPTITLE_FONTSIZE = 15
OVERLAY_ROI_LABEL_FONTSIZE = 10

# ------------------------------------------------------------
# 7. Utilities
# ------------------------------------------------------------
def parse_sample_tokens(sample_name):
    toks = str(sample_name).split("_")
    hydrophilic = toks[0] if len(toks) > 0 else "NA"
    hydrophobic = toks[1] if len(toks) > 1 else "NA"
    crosslinker = toks[2] if len(toks) > 2 else "NA"
    return hydrophilic, hydrophobic, crosslinker

def get_crosslinker_name(sample_name):
    return parse_sample_tokens(sample_name)[2]

def robust_levels(data, n_levels=8, percentile=98.5, mult=1.4):
    arr = np.asarray(data, dtype=float)
    arr = arr[np.isfinite(arr)]
    if arr.size == 0:
        return np.array([1.0])

    base = np.percentile(np.abs(arr), percentile)
    if base <= 0:
        mx = np.max(np.abs(arr))
        base = mx * 0.03 if mx > 0 else 1.0

    levels = np.array([base * (mult ** i) for i in range(n_levels)], dtype=float)
    levels = np.unique(levels[levels > 0])
    return levels if len(levels) > 0 else np.array([1.0])

def ppm_mesh(ppm_dim1, ppm_dim0):
    X, Y = np.meshgrid(ppm_dim1, ppm_dim0)
    return X, Y

def crop_2d_roi(ppm0, ppm1, data, w2_left, w2_right, w1_top, w1_bottom, pad_x=0.05, pad_y=0.05):
    dx = abs(w2_left - w2_right) * pad_x
    dy = abs(w1_top - w1_bottom) * pad_y

    x_lo = min(w2_left, w2_right) - dx
    x_hi = max(w2_left, w2_right) + dx
    y_lo = min(w1_top, w1_bottom) - dy
    y_hi = max(w1_top, w1_bottom) + dy

    mask_x = (ppm1 >= x_lo) & (ppm1 <= x_hi)
    mask_y = (ppm0 >= y_lo) & (ppm0 <= y_hi)

    if mask_x.sum() == 0 or mask_y.sum() == 0:
        return None, None, None

    cropped = data[np.ix_(mask_y, mask_x)]
    ppm0_c = ppm0[mask_y]
    ppm1_c = ppm1[mask_x]
    return ppm0_c, ppm1_c, cropped

def set_nmr_axes(ax):
    ax.invert_xaxis()
    ax.invert_yaxis()

def draw_hsqc_roi_boxes(ax, roi_table, label_only=None):
    for _, r in roi_table.iterrows():
        x1 = float(r["w2_left"])
        x2 = float(r["w2_right"])
        y1 = float(r["w1_top"])
        y2 = float(r["w1_bottom"])

        rect = patches.Rectangle(
            (min(x1, x2), min(y1, y2)),
            abs(x1 - x2),
            abs(y1 - y2),
            fill=False,
            edgecolor=ROI_BOX_COLOR,
            linewidth=0.9,
            zorder=6
        )
        ax.add_patch(rect)

        roi_label = str(r["ROI_Label"])
        if (label_only is None) or (roi_label in label_only):
            ax.text(
                (x1 + x2) / 2,
                max(y1, y2),
                roi_label,
                ha="center",
                va="bottom",
                fontsize=OVERLAY_ROI_LABEL_FONTSIZE,
                fontweight="bold",
                color=ROI_BOX_COLOR,
                zorder=7
            )

# ------------------------------------------------------------
# 8. Cache spectra + GLOBAL contour levels
# ------------------------------------------------------------
blank_dic, blank_data, blank_ppm0, blank_ppm1 = read_2d_ucsf(blank_path)

hsqc_cache = {
    "Blank": {
        "data": blank_data,
        "ppm0": blank_ppm0,
        "ppm1": blank_ppm1
    }
}

global_levels_cache = {}
global_levels_cache["Blank"] = robust_levels(
    blank_data,
    n_levels=N_LEVELS,
    percentile=BASE_NOISE_Q,
    mult=LEVEL_MULTIPLIER
)

for _, r in resolved_df.iterrows():
    if r["Resolved_File"] == "":
        continue

    no = int(r["No"])
    dic, data, ppm0, ppm1 = read_2d_ucsf(r["Resolved_File"])

    hsqc_cache[no] = {
        "data": data,
        "ppm0": ppm0,
        "ppm1": ppm1,
        "sample_name": r["Sample_Name"],
        "file": r["Resolved_File"]
    }

    global_levels_cache[no] = robust_levels(
        data,
        n_levels=N_LEVELS,
        percentile=BASE_NOISE_Q,
        mult=LEVEL_MULTIPLIER
    )

# ------------------------------------------------------------
# 9. HSQC overlay
#    Blank + 1–11, Blank + 12–23, Blank + 24–33, Blank + 34–43
# ------------------------------------------------------------
sample_entries = []
for _, r in resolved_df.sort_values("No", ascending=True).iterrows():
    if r["Resolved_File"] == "":
        continue
    sample_entries.append({
        "No": int(r["No"]),
        "Sample_Name": r["Sample_Name"]
    })

OVERLAY_GROUPS = [
    (1, 11),
    (12, 23),
    (24, 33),
    (34, 43),
]

sample_batches = []
for start_no, end_no in OVERLAY_GROUPS:
    batch = [
        x for x in sample_entries
        if start_no <= int(x["No"]) <= end_no
    ]
    if len(batch) > 0:
        sample_batches.append({
            "start": start_no,
            "end": end_no,
            "items": batch
        })

print("Number of overlay batches:", len(sample_batches))
for i, b in enumerate(sample_batches, start=1):
    print(f"Batch {i}: Blank + No.{b['start']}–{b['end']} -> {[x['No'] for x in b['items']]}")

from matplotlib.lines import Line2D

Xb, Yb = ppm_mesh(blank_ppm1, blank_ppm0)

overlay_output_files = []

for batch_idx, batch_info in enumerate(sample_batches, start=1):
    batch = batch_info["items"]
    start_no = batch_info["start"]
    end_no = batch_info["end"]

    fig, ax = plt.subplots(figsize=(12.5, 8))

    ax.contour(
        Xb, Yb, blank_data,
        levels=global_levels_cache["Blank"],
        colors=BLANK_COLOR,
        linewidths=1.0,
        alpha=0.95
    )

    handles = [Line2D([0], [0], color=BLANK_COLOR, lw=2)]
    labels = ["Blank"]

    for item in batch:
        no = item["No"]
        sample_name = item["Sample_Name"]
        cross = get_crosslinker_name(sample_name)
        color = CROSSLINKER_COLORS.get(cross, DEFAULT_COLOR)

        ppm0 = hsqc_cache[no]["ppm0"]
        ppm1 = hsqc_cache[no]["ppm1"]
        data = hsqc_cache[no]["data"]
        X, Y = ppm_mesh(ppm1, ppm0)

        ax.contour(
            X, Y, data,
            levels=global_levels_cache[no],
            colors=color,
            linewidths=0.55,
            alpha=0.82
        )

        handles.append(Line2D([0], [0], color=color, lw=2))
        labels.append(f"{no}. {sample_name}")

    # Overlay uses all ROIs
    draw_hsqc_roi_boxes(ax, roi_df_overlay, label_only=None)

    ax.set_xlabel(r"$^1$H chemical shift (ppm)", fontsize=AXIS_LABEL_FONTSIZE)
    ax.set_ylabel(r"$^{13}$C chemical shift (ppm)", fontsize=AXIS_LABEL_FONTSIZE)
    ax.set_title(
        f"HSQC overlay (Blank + No.{start_no}–{end_no})",
        fontsize=OVERLAY_TITLE_FONTSIZE,
        pad=10
    )

    set_nmr_axes(ax)

    ax.legend(
        handles,
        labels,
        frameon=False,
        fontsize=8,
        ncol=1,
        loc="upper left",
        bbox_to_anchor=(0.02, 0.98),
        borderaxespad=0.0,
        handlelength=2.0
    )

    out_png = f"hsqc_overlay_Blank_No{start_no:02d}-{end_no:02d}.png"
    out_pdf = f"hsqc_overlay_Blank_No{start_no:02d}-{end_no:02d}.pdf"

    plt.savefig(out_png, dpi=300, bbox_inches="tight")
    plt.savefig(out_pdf, bbox_inches="tight")
    plt.show()
    plt.close(fig)

    overlay_output_files.extend([out_png, out_pdf])
    print("Saved:", out_png, "/", out_pdf)

# ------------------------------------------------------------
# 10. ROI-based multi-window map
#     Uses ROI-excluded and re-numbered roi_df_multi
# ------------------------------------------------------------
plot_rows = [{"No": "Blank", "Label": "Blank"}]

for _, r in resolved_df.sort_values("No", ascending=True).iterrows():
    if r["Resolved_File"] == "":
        continue
    plot_rows.append({
        "No": int(r["No"]),
        "Label": f'{int(r["No"])}. {r["Sample_Name"]}'
    })

n_rows = len(plot_rows)
n_cols = len(roi_df_multi)

fig_w = max(14, COL_WIDTH * n_cols)
fig_h = max(9, ROW_HEIGHT * n_rows + 1.2)

fig, axes = plt.subplots(
    n_rows, n_cols,
    figsize=(fig_w, fig_h),
    squeeze=False,
    sharex=False,
    sharey=False,
    facecolor="white"
)

for row_idx, item in enumerate(plot_rows):
    row_key = item["No"]

    ppm0 = hsqc_cache[row_key]["ppm0"]
    ppm1 = hsqc_cache[row_key]["ppm1"]
    data = hsqc_cache[row_key]["data"]

    for col_idx, (_, rr) in enumerate(roi_df_multi.iterrows()):
        ax = axes[row_idx, col_idx]

        ppm0_c, ppm1_c, data_c = crop_2d_roi(
            ppm0, ppm1, data,
            rr["w2_left"], rr["w2_right"],
            rr["w1_top"], rr["w1_bottom"],
            pad_x=ROI_PAD_FRAC_X,
            pad_y=ROI_PAD_FRAC_Y
        )

        if data_c is None or data_c.size == 0:
            ax.axis("off")
            continue

        Xc, Yc = ppm_mesh(ppm1_c, ppm0_c)
        levels = global_levels_cache[row_key]

        ccolor = BLANK_COLOR if row_key == "Blank" else SAMPLE_CONTOUR_COLOR
        lw = 1.0 if row_key == "Blank" else 0.55
        alpha = 0.98 if row_key == "Blank" else 0.88

        ax.contour(
            Xc, Yc, data_c,
            levels=levels,
            colors=ccolor,
            linewidths=lw,
            alpha=alpha
        )

        ax.set_xlim(max(ppm1_c), min(ppm1_c))
        ax.set_ylim(max(ppm0_c), min(ppm0_c))

        ax.set_yticks([])
        ax.tick_params(axis="y", length=0)

        if row_idx == 0:
            ax.set_title(
                str(rr["ROI_Label"]),
                fontsize=ROI_TITLE_FONTSIZE,
                fontweight="bold",
                pad=6,
                color="black"
            )

        if col_idx == 0:
            ax.set_ylabel(
                item["Label"],
                fontsize=ROW_LABEL_FONTSIZE,
                rotation=0,
                ha="right",
                va="center",
                labelpad=26,
                color="black",
                fontweight="bold"
            )

        if row_idx == n_rows - 1:
            xticks = np.linspace(rr["w2_left"], rr["w2_right"], 3)
            ax.set_xticks(xticks)
            ax.set_xticklabels(
                [f"{x:.1f}" for x in xticks],
                fontsize=PPM_FONTSIZE,
                color="black"
            )
            ax.set_xlabel("ppm", fontsize=PPM_FONTSIZE, labelpad=2, color="black")
        else:
            ax.set_xticks([])

        for spine in ax.spines.values():
            spine.set_linewidth(0.6)
            spine.set_color("#666666")

fig.suptitle(
    "ROI-based multi-window HSQC map",
    fontsize=SUPTITLE_FONTSIZE,
    fontweight="bold",
    y=0.995
)

plt.tight_layout(rect=[0.06, 0.03, 1.00, 0.985])

multi_png = "hsqc_multiwindow_ROI_excluded_renumbered.png"
multi_pdf = "hsqc_multiwindow_ROI_excluded_renumbered.pdf"

plt.savefig(multi_png, dpi=300, bbox_inches="tight")
plt.savefig(multi_pdf, bbox_inches="tight")
plt.show()
plt.close(fig)

print("Saved:", multi_png, "/", multi_pdf)

# ------------------------------------------------------------
# 11. Save ROI mapping table
# ------------------------------------------------------------
roi_df_multi[[
    "Original_ROI_Label", "ROI_Label",
    "w2_left", "w2_right", "w1_top", "w1_bottom"
]].to_csv("multiwindow_ROI_relabeling_table.csv", index=False, encoding="utf-8-sig")

# ------------------------------------------------------------
# 12. Zip outputs
# ------------------------------------------------------------
output_files = [
    "resolved_hsqc_2d_files_code3_style.csv",
    "multiwindow_ROI_relabeling_table.csv",
    multi_png,
    multi_pdf,
]

output_files.extend(overlay_output_files)

zip_name = "hsqc_visualization_overlay4groups_multiwindow_ROIexcluded.zip"

with zipfile.ZipFile(zip_name, "w", zipfile.ZIP_DEFLATED) as zf:
    for fp in output_files:
        if os.path.exists(fp):
            zf.write(fp)

print("\nSaved files:")
for fp in output_files:
    if os.path.exists(fp):
        print(" -", fp)
print(" -", zip_name)

files.download(zip_name)
```
